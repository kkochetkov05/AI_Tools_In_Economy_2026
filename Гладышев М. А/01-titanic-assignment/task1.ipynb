```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('titanic.csv', index_col='PassengerId')
```

## 1. Какое количество мужчин и женщин ехало на корабле?
В качестве ответа приведите два числа через пробел.

```python
male_count = df['Sex'].value_counts()['male']
female_count = df['Sex'].value_counts()['female']
print(male_count, female_count)
```

# 2. Какой части пассажиров удалось выжить?
Посчитайте долю выживших пассажиров

```python
survived = round(df['Survived'].mean() * 100, 2)
print(f"{survived:.2f}%")
```

# 3. Какую долю пассажиры первого класса составляли среди всех пассажиров?
 Ответ приведите в процентах

```python
first_class = round((df['Pclass'] == 1).mean() * 100, 2)
print(f"{first_class:.2f}%")
```

# 4. Какого возраста были пассажиры?
Посчитайте среднее и медиану возраста пассажиров.

```python
age_clean = df['Age'].dropna()
age_mean = age_clean.mean()
age_median = age_clean.median()
print(age_mean, age_median)
```

# 5. Коррелируют ли число братьев/сестер с числом родителей/детей?
Посчитайте корреляцию Пирсона между признаками SibSp и Parch.

```python
corr_val = df['SibSp'].corr(df['Parch'])
print(corr_val)
```

# 6. Какое самое популярное женское имя на корабле?

```python
# Функция для извлечения имени
def extract_first_name(full_name, sex):
    if sex == 'female':
        if '(' in full_name:
            # Берем имя в скобках (часто для замужних)
            first = full_name.split('(')[1].split()[0].strip('"').strip(')')
        elif ' Miss.' in full_name:
            # Для Miss имя после точки и пробела
            first = full_name.split(' Miss. ')[1].split()[0]
        elif ' Mrs.' in full_name:
            first = full_name.split(' Mrs. ')[1].split()[0]
        else:
            return None
        # Убираем лишние символы
        first = first.strip('".').strip('()')
        return first
    return None

# Собираем имена для женщин
female_first_names = []
for name, sex in zip(df['Name'], df['Sex']):
    fname = extract_first_name(name, sex)
    if fname:
        female_first_names.append(fname)

# Подсчет популярности
name_counts = pd.Series(female_first_names).value_counts()
most_popular = name_counts.idxmax()
max_count = name_counts.max()

print(f"Ответ: {most_popular} ({max_count} раз)")

# Для проверки полного списка (необязательно)
print("Полный список женских имен:")
for name, count in name_counts.items():
    print(f"{name} {count}")
```
