```python
import pandas as pd
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import KFold, cross_val_score
from sklearn.preprocessing import scale
```

# 1. Загрузите выборку Wine по адресу https://archive.ics.uci.edu/ml/machine-learning-databases/wine/wine.data

```python
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/wine/wine.data"
wine_data = pd.read_csv(url, header=None)
```

# 2. Извлеките из данных признаки и классы. Класс записан в первом столбце (три варианта), признаки - в столбцах со второго по последний. Более подробно о сути признаков можно прочитать по адресу https://archive.ics.uci.edu/ml/datasets/Wine (см. также файл wine.names, приложенный к заданию)

```python
X_raw = wine_data.iloc[:, 1:]
y_raw = wine_data.iloc[:, 0]
```

# 3. Оценку качества необходимо провести методом кросс-валидации по 5 блокам (5-fold). Создайте генератор разбиений, который перемешивает выборку перед формированием блоков (shuffle=True). Для воспроизводимости результата, создавайте генератор KFold с фиксированным параметром random_state=42. В качестве меры качества используйте долю верных ответов (accuracy).

```python
cv_splitter = KFold(n_splits=5, shuffle=True, random_state=42)
```

# 4. Найдите точность классификации на кросс-валидации для метода k ближайших соседей (sklearn.neighbors.KNeighborsClassifier), при k от 1 до 50. При каком k получилось оптимальное качество? Чему оно равно (число в интервале от 0 до 1)? Данные результаты и будут ответами на вопросы 1 и 2.

```python
results_no_scale = []

for k in range(1, 51):
    knn_model = KNeighborsClassifier(n_neighbors=k)
    scores = cross_val_score(knn_model, X_raw, y_raw, cv=cv_splitter, scoring='accuracy')
    results_no_scale.append(scores.mean())

best_k_no_scale = np.argmax(results_no_scale) + 1
best_acc_no_scale = max(results_no_scale)

print(f"Лучшее k = {best_k_no_scale}")
print(f"Точность = {best_acc_no_scale:.2f}")
```

# 5. Произведите масштабирование признаков с помощью функции sklearn.preprocessing.scale. Снова найдите оптимальное k на кроссвалидации.

```python
X_scaled = scale(X_raw)
```

```python
results_with_scale = []

for k in range(1, 51):
    knn_model = KNeighborsClassifier(n_neighbors=k)
    scores = cross_val_score(knn_model, X_scaled, y_raw, cv=cv_splitter, scoring='accuracy')
    results_with_scale.append(scores.mean())

best_k_with_scale = np.argmax(results_with_scale) + 1
best_acc_with_scale = max(results_with_scale)
```

# 6. Какое значение k получилось оптимальным после приведения признаков к одному масштабу? Как изменилось значение качества? Приведите ответы на вопросы 3 и 4.

```python
print(f"Лучшее k = {best_k_with_scale}")
print(f"Точность = {best_acc_with_scale:.2f}")
```
