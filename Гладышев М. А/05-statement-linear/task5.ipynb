```python
import pandas as pd
import numpy as np
from sklearn.linear_model import Perceptron
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import StandardScaler
```

# 1. Загрузите обучающую и тестовую выборки из файлов perceptron train.csv и perceptron-test.csv. Целевая переменная записана в первом столбце, признаки - во втором и третьем.

```python
train_data = pd.read_csv('perceptron-train.csv', header=None)
test_data = pd.read_csv('perceptron-test.csv', header=None)
```

```python
X_train_raw = train_data.iloc[:, 1:]
y_train_raw = train_data.iloc[:, 0]

X_test_raw = test_data.iloc[:, 1:]
y_test_raw = test_data.iloc[:, 0]
```

# 2. Обучите персептрон со стандартными параметрами и random_state=241.

```python
perceptron_original = Perceptron(random_state=241)
perceptron_original.fit(X_train_raw, y_train_raw)
y_pred_original = perceptron_original.predict(X_test_raw)
```

# 3. Подсчитайте качество (долю правильно классифицированных объектов, accuracy) полученного классификатора на тестовой выборке.

```python
accuracy_before = accuracy_score(y_test_raw, y_pred_original)
print(f"Accuracy до нормализации: {accuracy_before:.3f}")
```

# 4. Нормализуйте обучающую и тестовую выборку с помощью класса StandardScaler.

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_raw)
X_test_scaled = scaler.transform(X_test_raw)
```

# 5. Обучите персептрон на новых выборках. Найдите долю правильных ответов на тестовой выборке.

```python
perceptron_scaled = Perceptron(random_state=241)
perceptron_scaled.fit(X_train_scaled, y_train_raw)

y_pred_scaled = perceptron_scaled.predict(X_test_scaled)

accuracy_after = accuracy_score(y_test_raw, y_pred_scaled)
print(f"Accuracy после нормализации: {accuracy_after:.3f}")
```

# 6. Найдите разность между качеством на тестовой выборке после нормализации и качеством до нее. Это число и будет ответом на задание.

```python
diff = accuracy_after - accuracy_before
diff_rounded = round(diff, 3)

print(f"{diff_rounded}")
```
