```python
import pandas as pd
import numpy as np
from sklearn.decomposition import PCA
```

# 1. Загрузите данные close_prices.csv. В этом файле приведены цены акций 30 компаний на закрытии торгов за каждый день периода.

```python
stock_prices = pd.read_csv('close_prices.csv')

X = stock_prices.iloc[:, 1:].values

company_names = stock_prices.columns[1:]
```

# 2. На загруженных данных обучите преобразование PCA с числом компоненты равным 10. Скольких компонент хватит, чтобы объяснить 90% дисперсии?

```python
pca_model = PCA(n_components=10)
pca_model.fit(X)

explained_variance_ratio = pca_model.explained_variance_ratio_
cumulative_variance = np.cumsum(explained_variance_ratio)

# Находим минимальное число компонент, где накопленная дисперсия >= 90%
n_components_90 = np.argmax(cumulative_variance >= 0.9) + 1
print(f"Число компонент для 90% дисперсии: {n_components_90}")
```

# 3. Примените построенное преобразование к исходным данным и возьмите значения первой компоненты.

```python
X_transformed = pca_model.transform(X)
first_component = X_transformed[:, 0]
```

# 4. Загрузите информацию об индексе Доу-Джонса из файла djia_prices.csv. Чему равна корреляция Пирсона между первой компонентой и индексом Доу-Джонса?

```python
djia_index = pd.read_csv('djia_index.csv')['^DJI'].values

# Корреляция Пирсона
correlation = np.corrcoef(first_component, djia_index)[0, 1]
print(f"Корреляция первой компоненты с DJIA: {correlation:.2f}")
```

# 5. Какая компания имеет наибольший вес в первой компоненте?

```python
weights_first_pc = pca_model.components_[0]

# Индекс компании с максимальным весом по модулю
max_weight_idx = np.argmax(np.abs(weights_first_pc))
top_company = company_names[max_weight_idx]
max_weight_value = weights_first_pc[max_weight_idx]

print(f"Компания с наибольшим весом в первой компоненте: {top_company}")
print(f"Значение веса: {max_weight_value:.4f}")
```
