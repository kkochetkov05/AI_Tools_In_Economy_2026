```python
import pandas as pd
import re
```

```python
df = pd.read_csv('titanic.csv', index_col='PassengerId')
df.head()
```

```python
# 1. Мужчины и женщины
male = (df['Sex'] == 'male').sum()
female = (df['Sex'] == 'female').sum()
ans1 = f"{male} {female}"
print(ans1)
```

```python
# 2. Доля выживших
survived = df['Survived'].mean() * 100
ans2 = f"{survived:.2f}"
print(ans2)
```

```python
# 3. Доля первого класса
first_class = (df['Pclass'] == 1).mean() * 100
ans3 = f"{first_class:.2f}"
print(ans3)
```

```python
# 4. Средний и медианный возраст
mean_age = df['Age'].mean()
median_age = df['Age'].median()
ans4 = f"{mean_age:.2f} {median_age:.2f}"
print(ans4)
```

```python
# 5. Корреляция Пирсона
corr = df['SibSp'].corr(df['Parch'])
ans5 = f"{corr:.2f}"
print(ans5)
```

```python
# 6. Самое популярное женское имя
def get_first_name(name):
    title = re.search(r',\s*(\w+\.)', name)
    if not title or title.group(1) not in ['Miss.', 'Mrs.', 'Ms.', 'Lady.', 'Countess.', 'Mme.', 'Mlle.']:
        return None
    bracket = re.search(r'\(([^)]+)\)', name)
    if bracket:
        return bracket.group(1).split()[0].strip()
    after_title = name.split(title.group(1))[1].strip()
    return after_title.split()[0].strip().rstrip(',.')

female_names = df[df['Sex'] == 'female']['Name'].apply(get_first_name).dropna().str.title()
common_name = female_names.value_counts().idxmax()
print(common_name)
```

```python
with open('1.txt', 'w') as f:
    f.write(ans1)

with open('2.txt', 'w') as f:
    f.write(ans2)

with open('3.txt', 'w') as f:
    f.write(ans3)

with open('4.txt', 'w') as f:
    f.write(ans4)

with open('5.txt', 'w') as f:
    f.write(ans5)

with open('6.txt', 'w') as f:
    f.write(common_name)
```
