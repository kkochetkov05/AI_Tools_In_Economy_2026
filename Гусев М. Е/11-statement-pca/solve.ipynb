```python
import pandas as pd
import numpy as np
from sklearn.decomposition import PCA
```

```python
# 1. Загрузка данных
prices = pd.read_csv('close_prices.csv')
djia = pd.read_csv('djia_index.csv')

# Первый столбец - дата, остальные - компании
X = prices.iloc[:, 1:].values
company_names = prices.columns[1:]

# 2. Обучение PCA
pca = PCA(n_components=10)
pca.fit(X)

# Определение количества компонент
cumsum = np.cumsum(pca.explained_variance_ratio_)
n_comp = np.argmax(cumsum >= 0.9) + 1  # индексация с 0

# 3. Применение преобразования и получение первой компоненты
X_pca = pca.transform(X)
first_component = X_pca[:, 0]

# 4. Корреляция с индексом Доу-Джонса
djia_values = djia['^DJI'].values if '^DJI' in djia.columns else djia.iloc[:, 1].values
corr = np.corrcoef(first_component, djia_values)[0, 1]
corr = round(corr, 2)

# 5. Компания с наибольшим весом в первой компоненте
max_weight_idx = np.argmax(np.abs(pca.components_[0]))
company_with_max_weight = company_names[max_weight_idx]


print(f"1. Компонент для 90% дисперсии: {n_comp}")
print(f"2. Корреляция: {corr}")
print(f"3. Компания с наибольшим весом: {company_with_max_weight}")
```

```python
with open('answer1.txt', 'w') as f:
    f.write(str(n_comp))

with open('answer2.txt', 'w') as f:
    f.write(str(corr))

with open('answer3.txt', 'w') as f:
    f.write(company_with_max_weight)
```
