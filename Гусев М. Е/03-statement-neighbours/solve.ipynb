```python
import numpy as np
import pandas as pd
from sklearn.model_selection import KFold, cross_val_score
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import scale
```

```python
# 1. Загрузка данных
data = pd.read_csv("wine.data", header=None)
```

```python
# 2. Разделение на признаки и классы
X = data.iloc[:, 1:].values
y = data.iloc[:, 0].values

# 3. Генератор разбиений
kf = KFold(n_splits=5, shuffle=True, random_state=42)

# 4. Подбор k от 1 до 50 без масштабирования
scores = []
for k in range(1, 51):
    knn = KNeighborsClassifier(n_neighbors=k)
    cv_scores = cross_val_score(knn, X, y, cv=kf)
    scores.append(np.mean(cv_scores))

best_k_raw = range(1, 51)[np.argmax(scores)]
best_score_raw = round(np.max(scores), 2)

# 5. Масштабирование признаков
X_scaled = scale(X)

# 6. После масштабирования
scores_scaled = []
for k in range(1, 51):
    knn = KNeighborsClassifier(n_neighbors=k)
    cv_scores = cross_val_score(knn, X_scaled, y, cv=kf)
    scores_scaled.append(np.mean(cv_scores))

best_k_scaled = range(1, 51)[np.argmax(scores_scaled)]
best_score_scaled = round(np.max(scores_scaled), 2)


print(f"1. Оптимальное k (без масштабирования): {best_k_raw}")
print(f"2. Качество до масштабирования: {best_score_raw}")
print(f"3. Оптимальное k (после масштабирования): {best_k_scaled}")
print(f"4. Качество после масштабирования: {best_score_scaled}")
```

```python
with open("1.txt", "w") as f:
    f.write(str(best_k_raw))
with open("2.txt", "w") as f:
    f.write(str(best_score_raw))
with open("3.txt", "w") as f:
    f.write(str(best_k_scaled))
with open("4.txt", "w") as f:
    f.write(str(best_score_scaled))
```
