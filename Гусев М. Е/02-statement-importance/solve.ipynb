```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv('titanic.csv')
```

```python
# Оставляем 4 признака и целевую переменную
features = ['Pclass', 'Fare', 'Age', 'Sex']
target = 'Survived'

# Преобразуем пол в число 0 или 1
le = LabelEncoder()
df['Sex'] = le.fit_transform(df['Sex'])

# Удаляем строки с пропусками
data = df[features + [target]].dropna()

# Обучаем дерево
clf = DecisionTreeClassifier(random_state=241)
clf.fit(data[features], data[target])

# Получаем важности и находим два самых важных признака
importances = clf.feature_importances_
feat_imp = dict(zip(features, importances))
top2 = sorted(feat_imp, key=feat_imp.get, reverse=True)[:2]

ans = ','.join(top2)
print(ans)
```

```python
with open('1.txt', 'w') as f:
    f.write(ans)
```
