```python
import numpy as np
import pandas as pd
from sklearn.metrics import roc_auc_score
```

```python
data = pd.read_csv('data-logistic.csv', header=None)
X = data.iloc[:, 1:].values
y = data.iloc[:, 0].values
```

```python
def sigmoid(t):
    return 1 / (1 + np.exp(-t))

def train_logistic(X, y, C=0.0, lr=0.1, max_iter=10000, tol=1e-5):
    n_samples, n_features = X.shape
    w = np.zeros(n_features)
    for _ in range(max_iter):

        # Просто знаменатель
        factor = (1 / (1 + np.exp(y * np.dot(X, w))))

        # Градиент функционала потерь
        grad_loss = (X.T @ (y * factor)) / n_samples

        # Обновление весов
        w_new = w + lr * (grad_loss - C * w)

        if np.linalg.norm(w_new - w) <= tol:
            w = w_new
            break
        w = w_new
    return w
```

```python
# Обучение без/с L2-регуляризацией
w_no_reg = train_logistic(X, y, C=0.0)
w_reg = train_logistic(X, y, C=10.0)

# Вычисление вероятностей (сигмоида от линейной комбинации)
prob_no_reg = sigmoid(np.dot(X, w_no_reg))
prob_reg = sigmoid(np.dot(X, w_reg))

# Преобразование меток в бинарные (-1/1 -> 0/1)
y_bin = (y == 1).astype(int)

# AUC-ROC
auc_no_reg = roc_auc_score(y_bin, prob_no_reg)
auc_reg = roc_auc_score(y_bin, prob_reg)

answer = f"{round(auc_no_reg, 3)} {round(auc_reg, 3)}"

print(answer)
```

```python
with open("answer.txt", "w") as f:
    f.write(answer)
```
