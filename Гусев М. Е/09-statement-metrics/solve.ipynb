```python
import pandas as pd
import numpy as np
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, precision_recall_curve)
```

```python
df_clf = pd.read_csv('classification.csv')
y_true = df_clf['true']
y_pred = df_clf['pred']

TP = ((y_true == 1) & (y_pred == 1)).sum()
FP = ((y_true == 0) & (y_pred == 1)).sum()
FN = ((y_true == 1) & (y_pred == 0)).sum()
TN = ((y_true == 0) & (y_pred == 0)).sum()

acc = round(accuracy_score(y_true, y_pred), 2)
prec = round(precision_score(y_true, y_pred), 2)
rec = round(recall_score(y_true, y_pred), 2)
f1 = round(f1_score(y_true, y_pred), 2)
```

```python
df_scores = pd.read_csv('scores.csv')
y_true_scores = df_scores['true']
score_cols = ['score_logreg', 'score_svm', 'score_knn', 'score_tree']

auc_dict = {}
for col in score_cols:
    auc = roc_auc_score(y_true_scores, df_scores[col])
    auc_dict[col] = auc

best_auc_col = max(auc_dict, key=auc_dict.get)


# Максимальная precision при recall >= 0.7
best_classifier = None
best_precision = -1.0

for col in score_cols:
    prec_curve, recall_curve, _ = precision_recall_curve(y_true_scores, df_scores[col])
    mask = recall_curve >= 0.7
    if np.any(mask):
        max_prec = np.max(prec_curve[mask])
        if max_prec > best_precision:
            best_precision = max_prec
            best_classifier = col

print(f"TP={TP}, FP={FP}, FN={FN}, TN={TN}")
print(f"accuracy={acc}, precision={prec}, recall={rec}, F1={f1}")
print(best_auc_col)
print(f"{best_classifier} {round(best_precision, 2)}")
```

```python
with open("1.txt", "w") as f:
    f.write(f"{TP} {FP} {FN} {TN}")

with open("2.txt", "w") as f:
    f.write(f"{acc} {prec} {rec} {f1}")

with open("3.txt", "w") as f:
    f.write(best_auc_col)

with open("4.txt", "w") as f:
    f.write(f"{best_classifier} {round(best_precision, 2)}")
```
