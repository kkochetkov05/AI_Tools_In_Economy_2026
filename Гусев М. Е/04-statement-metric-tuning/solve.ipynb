```python
import numpy as np
import pandas as pd
from sklearn.model_selection import KFold, cross_val_score
from sklearn.neighbors import KNeighborsRegressor
from sklearn.preprocessing import scale
```

```python
data_url = "http://lib.stat.cmu.edu/datasets/boston"
raw_df = pd.read_csv(data_url, sep=r"\s+", skiprows=22, header=None)
X = np.hstack([raw_df.values[::2, :], raw_df.values[1::2, :2]])
y = raw_df.values[1::2, 2]
```

```python
# Масштабируем признаки
X_scaled = scale(X)

# Генератор кросс-валидации
kf = KFold(n_splits=5, shuffle=True, random_state=42)

best_score = -np.inf
best_p = None

for p in np.linspace(1, 10, 200):
    knn = KNeighborsRegressor(n_neighbors=5, weights='distance', metric='minkowski', p=p)
    scores = cross_val_score(knn, X_scaled, y, cv=kf, scoring='neg_mean_squared_error')
    mean_score = np.mean(scores)
    if mean_score > best_score:
        best_score = mean_score
        best_p = p


print(f"Оптимальное p: {round(best_p, 1)}")
print(f"Лучшее значение neg_mean_squared_error: {best_score:.4f}")
```

```python
with open("answer.txt", "w") as f:
    f.write(str(round(best_p, 1)))
```
