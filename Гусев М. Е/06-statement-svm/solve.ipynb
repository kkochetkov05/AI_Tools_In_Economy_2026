```python
import pandas as pd
from sklearn.svm import SVC
```

```python
data = pd.read_csv("svm-data.csv", header=None)
X = data.iloc[:, 1:].values
y = data.iloc[:, 0].values
```

```python
# Обучаем SVM с линейным ядром
clf = SVC(kernel='linear', C=100000, random_state=241)
clf.fit(X, y)

# Получаем индексы опорных векторов
support_indices = clf.support_ + 1
support_indices_sorted = sorted(support_indices)

answer = ",".join(map(str, support_indices_sorted))


print("Опорные объекты:", answer)
```

```python
with open("answer.txt", "w") as f:
    f.write(answer)
```
