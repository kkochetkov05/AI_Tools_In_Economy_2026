```python
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import KFold, cross_val_score
from sklearn.metrics import r2_score
```

1. Загрузите данные из файла abalone.csv. Это датасет, в котором требуется предсказать возраст ракушки (число колец) по физическим измерениям.

```python
data = pd.read_csv('abalone.csv')
print(data.shape)
data.head()
```

2. Преобразуйте признак Sex в числовой: значение F должно перейти в -1, I — в 0, M — в 1.

```python
data['Sex'] = data['Sex'].map(lambda x: 1 if x == 'M' else (-1 if x == 'F' else 0))
print(data['Sex'].value_counts())
```

3. Разделите содержимое файлов на признаки и целевую переменную. В последнем столбце записана целевая переменная (Rings), в остальных — признаки.

```python
X = data.iloc[:, :-1].values
y = data.iloc[:, -1].values
print('X shape:', X.shape, '  y shape:', y.shape)
```

4. Обучите случайный лес (RandomForestRegressor) с различным числом деревьев: от 1 до 50 (random_state=1). Для каждого варианта оцените качество на кросс-валидации по 5 блокам (random_state=1, shuffle=True). В качестве меры качества используйте R2 (r2_score).

```python
kf = KFold(n_splits=5, shuffle=True, random_state=1)

scores = []
for n_trees in range(1, 51):
    clf = RandomForestRegressor(n_estimators=n_trees, random_state=1)
    cv_scores = cross_val_score(clf, X, y, cv=kf, scoring='r2')
    scores.append(cv_scores.mean())

for i, s in enumerate(scores, 1):
    print(f'n_trees={i:2d}: R2={s:.4f}')
```

5. Определите, при каком минимальном количестве деревьев случайный лес показывает качество на кросс-валидации выше 0.52. Это количество и будет ответом на задание.

```python
for n_trees, score in enumerate(scores, 1):
    if score > 0.52:
        print(f'Минимальное количество деревьев для R2 > 0.52: {n_trees}')
        print(f'R2 = {score:.4f}')
        break
```

6. Обратите внимание на изменение качества по мере роста числа деревьев. Ухудшается ли оно?

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 4))
plt.plot(range(1, 51), scores, marker='o', markersize=3)
plt.axhline(y=0.52, color='r', linestyle='--', label='0.52')
plt.xlabel('Количество деревьев')
plt.ylabel('R2 (кросс-валидация)')
plt.title('Качество случайного леса в зависимости от числа деревьев')
plt.legend()
plt.grid(True)
plt.show()

print('Качество не ухудшается — случайный лес не переобучается по мере добавления деревьев.')
```
