```python
import pandas as pd
import numpy as np
from sklearn.decomposition import PCA
```

1. Загрузите данные close_prices.csv. В этом файле приведены цены акций 30 компаний на закрытии торгов за каждый день периода.

```python
prices = pd.read_csv('close_prices.csv')
X = prices.drop('date', axis=1).values
print('Shape:', X.shape)
prices.head()
```

2. На загруженных данных обучите преобразование PCA с числом компонент равным 10. Скольких компонент хватит, чтобы объяснить 90% дисперсии?

```python
pca = PCA(n_components=10)
pca.fit(X)

print('Объяснённая дисперсия по компонентам:')
for i, v in enumerate(pca.explained_variance_ratio_, 1):
    print(f'  Компонента {i}: {v:.4f}')

cum_var = np.cumsum(pca.explained_variance_ratio_)
n_components_90 = int(np.searchsorted(cum_var, 0.9) + 1)
print(f'\nКумулятивная дисперсия: {cum_var}')
print(f'Компонент для объяснения 90% дисперсии: {n_components_90}')
```

3. Примените построенное преобразование к исходным данным и возьмите значения первой компоненты.

```python
X_pca = pca.transform(X)
first_component = X_pca[:, 0]
print('Первая компонента (первые 5 значений):', first_component[:5])
```

4. Загрузите информацию об индексе Доу-Джонса из файла djia_index.csv. Чему равна корреляция Пирсона между первой компонентой и индексом Доу-Джонса?

```python
djia = pd.read_csv('djia_index.csv')
dji_values = djia['^DJI'].values

corr = np.corrcoef(first_component, dji_values)[0, 1]
print(f'Корреляция Пирсона между 1-й компонентой и DJIA: {round(corr, 2)}')
```

5. Какая компания имеет наибольший вес в первой компоненте?

```python
feature_names = prices.drop('date', axis=1).columns
weights = pca.components_[0]

best_company = feature_names[np.argmax(np.abs(weights))]
print('Компания с наибольшим весом в 1-й компоненте:', best_company)
print(f'Вес: {weights[np.argmax(np.abs(weights))]:.4f}')
```
