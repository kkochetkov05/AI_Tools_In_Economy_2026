```python
import pandas as pd
data = pd.read_csv('titanic.csv', index_col='PassengerId')
```

1. Какое количество мужчин и женщин ехало на корабле? В качестве ответа приведите два числа через пробел.

```python
print(data["Sex"].value_counts()["male"], data["Sex"].value_counts()["female"])
```

2. Какой части пассажиров удалось выжить? Посчитайте долю вы-
живших пассажиров. Ответ приведите в процентах (число в ин-
тервале от 0 до 100, знак процента не нужен), округлив до двух
знаков.

```python
all_passengers = len(data)
survived_passengers = data['Survived'].value_counts()[1]
print(round(survived_passengers / all_passengers * 100, 2))
```

3. Какую долю пассажиры первого класса составляли среди всех пассажиров? Ответ приведите в процентах (число в интервале от 0 до 100, знак процента не нужен), округлив до двух знаков.


```python
all_passengers = len(data)
first_class_passengers = data['Pclass'].value_counts()[1]
print(round(first_class_passengers / all_passengers * 100, 2))
```

4. Какого возраста были пассажиры? Посчитайте среднее и медиа-
ну возраста пассажиров. Посчитайте среднее и медиану возраста
пассажиров. В качестве ответа приведите два числа через пробел.

```python
mean_age = data['Age'].mean()
median_age = data['Age'].median()
print(round(mean_age, 2), round(median_age, 2))
```

5. Коррелируют ли число братьев/сестер с числом родителей/детей?
Посчитайте корреляцию Пирсона между признаками SibSp и Parch.

```python
pearson_corr = data["SibSp"].corr(data["Parch"], method="pearson")
print(round(pearson_corr, 2))
```

6. Какое самое популярное женское имя на корабле? Извлеките из
полного имени пассажира (колонка Name) его личное имя (First
Name). Это задание - типичный пример того, с чем сталкивает-
ся специалист по анализу данных. Данные очень разнородные и
шумные, но из них требуется извлечь необходимую информацию.
Попробуйте вручную разобрать несколько значений столбца Name
и выработать правило для извлечения имен, а также разделения
их на женские и мужские

```python
def extract_first_name(full_name):
    if '(' in full_name:
        return full_name.split('(')[1].split(')')[0].split()[0]
    else:
        return full_name.split(',')[1].split('.')[1].strip().split()[0]

data['FirstName'] = data['Name'].apply(extract_first_name)

female_names = data[data['Sex'] == 'female']['FirstName']
print(female_names.value_counts().idxmax())
```
