```python
import pandas as pd
import numpy as np
from sklearn.metrics import roc_auc_score
```

```python
data = pd.read_csv('data-logistic.csv')
y = data.iloc[:, 0].values
X = data.iloc[:, 1:].values
```

3. Реализуйте градиентный спуск для обычной и L2-регуляризованной
(с коэффициентом регуляризации 10) логистической регрессии. Ис-
пользуйте длину шага k=0.1. В качестве начального приближения
используйте вектор (0, 0).

```python
def gradient_descent(X, y, C=0, k=0.1, max_iter=10000, tol=1e-5):
    """
    w1 := w1 + k * (1/l) * Σ y_i * x_i1 * (1 - 1/(1+exp(-y_i*(w1*x_i1 + w2*x_i2)))) - k*C*w1
    w2 := w2 + k * (1/l) * Σ y_i * x_i2 * (1 - 1/(1+exp(-y_i*(w1*x_i1 + w2*x_i2)))) - k*C*w2
    """
    n_features = X.shape[1]
    w = np.zeros(n_features)

    for iteration in range(max_iter):
        z = y * np.dot(X, w)
        sigmoid_term = 1 / (1 + np.exp(-z))
        gradient = np.array([np.mean(y * X[:, j] * (1 - sigmoid_term)) for j in range(n_features)])
        w_new = w + k * gradient - k * C * w
        if np.linalg.norm(w_new - w) < tol:
            return w_new
        w = w_new

    return w
```

4. Запустите градиентный спуск и доведите до сходимости (евклидово
расстояние между векторами весов на соседних итерациях долж-
но быть не больше 1e-5). Рекомендуется ограничить сверху число
итераций десятью тысячами.

```python
w_no_reg = gradient_descent(X, y, C=0.0, k=0.1)
w_reg    = gradient_descent(X, y, C=10.0, k=0.1)
print(f"Веса без регуляризации: {w_no_reg}")
print(f"Веса с регуляризацией:  {w_reg}")
```

5. Посчитайте AUC-ROC для каждого варианта. Какое из двух значений оказалось больше?

```python
def predict_probability(X, w):
    return 1 / (1 + np.exp(-np.dot(X, w)))

auc_no_reg = roc_auc_score(y, predict_probability(X, w_no_reg))
auc_reg    = roc_auc_score(y, predict_probability(X, w_reg))
print(round(auc_no_reg, 3), round(auc_reg, 3))
```
