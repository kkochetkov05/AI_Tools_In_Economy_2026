6. Какой классификатор достигает наибольшей точности (Precision)
при полноте (Recall) не менее 70% (укажите название столбца с от
ветами этого классификатора)? Какое значение точности при этом
получается?

```python
precision_recall_scores = {}
for i in scores.columns[1:]:
    curve = precision_recall_curve(scores['true'], scores[i])
    df = pd.DataFrame({'precision': curve[0], 'recall': curve[1]})
    precision_recall_scores[i] = df[df['recall'] >= 0.7]['precision'].max()
answer6 = pd.Series(precision_recall_scores).idxmax()
answer7 = pd.Series(precision_recall_scores).max()
print(answer6, answer7)
```
