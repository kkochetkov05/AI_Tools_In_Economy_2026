```python
import pandas as pd
import numpy as np
from sklearn.metrics import roc_auc_score
```

1. Загрузите данные из файла data-logistic.csv. Это двумерная выборка, целевая переменная на которой принимает значения-1 или 1.

```python
data = pd.read_csv('../data/data-logistic.csv', header=None)
X = data[[1, 2]].to_numpy()
y = data[0].to_numpy()
```

2. Убедитесь, что выше выписаны правильные формулы для градиентного спуска. Обратите внимание, что мы используем полноценный градиентный спуск, а не его стохастический вариант!

3. Реализуйте градиентный спуск для обычной и L2-регуляризованной
(с коэффициентом регуляризации 10) логистической регрессии. Ис
пользуйте длину шага k=0.1. В качестве начального приближения
используйте вектор (0, 0).

```python
def step_w(X, y, w, k, C):
    s = np.zeros([2])
    l = len(y)
    for i in range(l):
        s += y[i] * X[i] * (1 - 1 / (1 + np.exp(-y[i] * np.dot(w, X[i]))))
    return k / l  * s - k * C * w

def grad(X, y, w, k, C):
    wp = np.zeros([2])
    i = 0
    while True:
        wp = w.copy()
        w += step_w(X, y, w, k, C)
        i += 1
        if i > 10000 or np.linalg.norm(w - wp) <= 1e-5:
            break
    return i, w
```

4. Запустите градиентный спуск и доведите до сходимости (евклидово
расстояние между векторами весов на соседних итерациях долж
но быть не больше 1e-5). Рекомендуется ограничить сверху число
итераций десятью тысячами.

```python
_, w1 = grad(X, y, np.zeros([2]), 0.1, 10)
```

5. Какое значение принимает AUC-ROC на обучении без регуляризации и при ее использовании? Эти величины будут ответом на
задание. В качестве ответа приведите два числа через пробел. Обратите внимание, что на вход функции roc_auc_score нужно подавать оценки вероятностей, подсчитанные обученным алгоритмом.
Для этого воспользуйтесь сигмоидной функцией: a(x) = 1/(1 +
exp(−w1x1 −w2x2)).

```python
a1 = 1 / (1 + np.exp(-X @ w1.T))
answer1 = roc_auc_score(y, a1)
_, w2 = grad(X, y, np.zeros([2]), 0.1, 0)
a2 = 1 / (1 + np.exp(-X @ w2.T))
answer2 = roc_auc_score(y, a2)
print(answer1, answer2)
```
