```python
import numpy as np
from sklearn.tree import DecisionTreeClassifier
import pandas as pd
data = pd.read_csv("titanic.csv", index_col="PassengerId", sep = ',', encoding = 'cp1251', skipinitialspace = True)
count = data['Sex'].value_counts()
new_data = data.drop(['Name','SibSp','Parch', 'Ticket','Cabin','Embarked'], axis = 1)
new_data = new_data.dropna()
y = new_data['Survived']
new_data = new_data.drop(['Survived'], axis = 1)
new_data['Sex'] = new_data['Sex'].map({'male': 0, 'female': 1})
features = ['Pclass', 'Fare', 'Age', 'Sex']
clf = DecisionTreeClassifier(random_state=241)
clf.fit(new_data, y)
importances = clf.feature_importances_
feature_importances = list(zip(features, importances))
feature_importances = sorted(feature_importances, key=lambda x: x[1], reverse=True)
top_two = feature_importances[:2]
answer = ','.join([feat for feat, imp in top_two])
print(answer)
```
