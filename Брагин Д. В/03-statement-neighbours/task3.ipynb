Оптимальное k: 29; Лучшее качество: 0.98 (со стандартизацией)

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import KFold
from sklearn.preprocessing import scale
from tools import find_best_k

df = pd.read_csv('wine.data', header=None)

y = df[0]
X = df.drop(0, axis=1) # удаляем идекс ноль, axis=1 значит по колонкам axis=0 по строкам

kf = KFold(n_splits=5, shuffle=True, random_state=42)


best_k, best_score = find_best_k(X, y, kf)


print(f"Задание 1 (лучшее k): {best_k}")
print(f"Задание 2 (лучшее качество): {round(best_score, 2)}")
```

```python
X_scaled = scale(X) # z-стандартизация, отнимаем мат ожидание делим на стд отклонение

# 6. Снова ищем оптимальное k
best_k_scaled, best_score_scaled = find_best_k(X_scaled, y, kf)

print(f"Задание 3 (лучшее k после масштабирования): {best_k_scaled}")
print(f"Задание 4 (лучшее качество после масштабирования): {round(best_score_scaled, 2)}")

```
