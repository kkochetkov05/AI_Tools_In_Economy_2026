```python
import pandas as pd
import re


# Загрузка данных
df = pd.read_csv('titanic.csv', index_col='PassengerId')
```

### 1. Какое количество мужчин и женщин ехало на корабле? В качестве
ответа приведите два числа через пробел.

```python

male_cnt = sum(df['Sex'] == 'male')
female_cnt = sum(df['Sex'] == 'female')
print(f"{male_cnt} {female_cnt}")
```

### 2. Какой части пассажиров удалось выжить?

```python
survived_pct = df['Survived'].mean() * 100
print(f"{survived_pct:.2f}%")
```

###  3. Какую долю пассажиры первого класса составляли среди всех пассажиров?

```python
first_class_pct = (df['Pclass'] == 1).mean() * 100
print(f"{first_class_pct:.2f}%")
```

### 5. Коррелируют ли число братьев/сестер с числом родителей/детей?

### 6. Какое самое популярное женское имя на корабле? 
Извлеките из полного имени пассажира (колонка Name) его личное имя (First
Name)

```python
# ========== 6. Самое популярное женское имя ==========
def extract_female_first_name(name):
    """
    Извлекает личное имя женщины из полной записи Name.
    Стратегия:
      1. Если в строке есть круглые скобки, берём содержимое внутри них.
      2. Иначе берём слово после титула (Miss. / Mrs. / Ms. / Mme. / Mlle.),
         отбрасывая возможную запятую или кавычки.
    """
    # Если есть скобки — содержимое в них содержит женское имя
    if '(' in name and ')' in name:
        inner = name.split('(')[1].split(')')[0]
        # Первое слово в скобках — это имя
        first = inner.split()[0]
        return first.strip(',. "')
    else:
        # Нет скобок — ищем после титула
        # Титулы для женщин: Miss., Mrs., Ms., Mme., Mlle.
        match = re.search(r'\b(Miss|Mrs|Ms|Mme|Mlle)\.\s+([A-Za-z]+)', name)
        if match:
            return match.group(2).strip(',. "')
        # Если не нашли, возвращаем None
        return None

# Применяем только для женщин
female_df = df[df['Sex'] == 'female'].copy()
female_df['First'] = female_df['Name'].apply(extract_female_first_name)

# Убираем пропуски (не удалось извлечь)
valid_names = female_df['First'].dropna()

# Самое часто встречающееся имя
most_common_name = valid_names.mode()[0] if not valid_names.empty else "None"
print(most_common_name)
```
