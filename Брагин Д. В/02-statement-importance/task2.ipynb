## 02-statement-importance
Ответ: Fare,Sex

```python
import pandas as pd
import numpy as np
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import LabelEncoder

# 1. Загрузка данных
df = pd.read_csv('titanic.csv', index_col='PassengerId')

# 2. Отбор признаков и целевой переменной
features = ['Pclass', 'Fare', 'Age', 'Sex']
X = df[features].copy()
y = df['Survived']

# 3. Преобразование пола в числовой
le = LabelEncoder()
X['Sex'] = le.fit_transform(X['Sex'])  

# 4. Удаление строк с пропусками
X = X.dropna()
y = y.loc[X.index]
```

```python

# 5. Обучение дерева
clf = DecisionTreeClassifier(random_state=241)
clf.fit(X, y)

# 6. Важности признаков
importances = clf.feature_importances_
sorted_idx = np.argsort(importances)[::-1]
print(sorted_idx, importances)
top2 = [features[i] for i in sorted_idx[:2]]
answer = ','.join(top2)


print("Два признака с наибольшей важностью:", answer)
```
