```python
import pandas as pd

# Загружаем обучающую и тестовую выборки
train_data = pd.read_csv("perceptron-train.csv", header=None)
test_data = pd.read_csv("perceptron-test.csv", header=None)

# В первом столбце (индекс 0) — целевая переменная
y_train = train_data.iloc[:, 0].values
X_train = train_data.iloc[:, 1:].values   # столбцы 1 и 2

y_test = test_data.iloc[:, 0].values
X_test = test_data.iloc[:, 1:].values

print("Размерность признаков (train):", X_train.shape)
print("Размерность целевой переменной (train):", y_train.shape)

```

```python
import numpy as np
from sklearn.linear_model import Perceptron
from sklearn.metrics import accuracy_score


# Создаём персептрон с random_state=241 для воспроизводимости
perceptron = Perceptron(random_state=241)

# Обучаем на исходных (ненормализованных) данных
perceptron.fit(X_train, y_train)

# Предсказываем на тестовой выборке
y_pred_raw = perceptron.predict(X_test)

# Считаем accuracy (долю правильных ответов)
accuracy_raw = accuracy_score(y_test, y_pred_raw)
print(f"Accuracy без нормализации: {accuracy_raw:.3f}")
```

```python
from sklearn.preprocessing import StandardScaler
# Создаём scaler
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

# Трансформируем тестовые данные, используя параметры, вычисленные на train
X_test_scaled = scaler.transform(X_test)

# Обучаем новый персептрон на нормализованных данных
perceptron_scaled = Perceptron(random_state=241)
perceptron_scaled.fit(X_train_scaled, y_train)

# Предсказываем
y_pred_scaled = perceptron_scaled.predict(X_test_scaled)

# Считаем accuracy
accuracy_scaled = accuracy_score(y_test, y_pred_scaled)
print(f"Accuracy с нормализацией: {accuracy_scaled:.3f}")
```

```python
difference = accuracy_scaled - accuracy_raw
print(f"Разность: {difference:.3f} - Ответ")
```
