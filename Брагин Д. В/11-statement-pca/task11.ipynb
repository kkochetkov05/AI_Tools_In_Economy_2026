## 11-statement-pca

Корреляция Пирсона: 0.91

```python
import pandas as pd
import numpy as np
from sklearn.decomposition import PCA

# ------------------------------------------------------------
# 1. Загрузка данных: предполагаем, что первый столбец - дата или индекс
# ------------------------------------------------------------
close_prices = pd.read_csv('close_prices.csv', index_col=0)   # дата становится индексом
djia = pd.read_csv('djia_index.csv', index_col=0)           # дата становится индексом

# Убедимся, что close_prices содержит только цены (числовые столбцы)
# Если какой-то столбец не числовой, его можно исключить
# Но в задании все столбцы должны быть ценами

# ------------------------------------------------------------
# 2. PCA на 10 компонент
# ------------------------------------------------------------
pca = PCA(n_components=10)
pca.fit(close_prices)

# ------------------------------------------------------------
# 3. Сколько компонент нужно для 90% дисперсии?
# ------------------------------------------------------------
explained_variance = pca.explained_variance_ratio_
cumulative = np.cumsum(explained_variance)
n_components_90 = np.argmax(cumulative >= 0.9) + 1
print(f"Число компонент для 90% дисперсии: {n_components_90}")

# ------------------------------------------------------------
# 4. Первая компонента и корреляция с индексом Доу-Джонса
# ------------------------------------------------------------
first_component = pca.transform(close_prices)[:, 0]
# Индекс Доу-Джонса: обычно столбец '^DJI' или 'DJI'
if '^DJI' in djia.columns:
    djia_values = djia['^DJI']
else:
    # возможно, столбец называется 'DJI' или это первая колонка
    djia_values = djia.iloc[:, 0]

correlation = np.corrcoef(first_component, djia_values)[0, 1]
print(f"Корреляция первой компоненты с индексом: {correlation:.4f}")

# ------------------------------------------------------------
# 5. Компания с наибольшим весом в первой компоненте
# ------------------------------------------------------------
weights = pca.components_[0]
# Найдем индекс максимального по модулю веса (вклад в компоненту)
max_weight_idx = np.argmax(np.abs(weights))
company = close_prices.columns[max_weight_idx]
print(f"Компания с наибольшим весом: {company}")

# Для вывода ответов (в одну строку или отдельно) - по заданию нужно три ответа:
print(f"{n_components_90}, Корреляция пирсона: {correlation:.2f} {company}")
```
