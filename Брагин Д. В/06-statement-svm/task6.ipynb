## 06-statement-svm

Ответ: 4,5,10

```python
import pandas as pd


svm_data = pd.read_csv("svm-data.csv", header=None)

y = svm_data.iloc[:, 0].values
X = svm_data.iloc[:, 1:].values
```

```python
from sklearn.svm import SVC

clf = SVC(kernel='linear', C=100000, random_state=241)
clf.fit(X, y)

# 3. Индексы опорных векторов
# clf.support_ возвращает индексы в исходной выборке (начиная с 0)
support_indices = clf.support_

# Переводим в нумерацию с 1 и сортируем
support_numbers = sorted([i + 1 for i in support_indices])

answer = ",".join(map(str, support_numbers))
print("Ответ:", answer)
```
