```python
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import KFold
from sklearn.metrics import r2_score

data = pd.read_csv('abalone.csv')

# ------------------------------------------------------------
# Преобразование признака Sex: F -> -1, I -> 0, M -> 1

data['Sex'] = data['Sex'].map(lambda x: 1 if x == 'M' else (-1 if x == 'F' else 0))

# ------------------------------------------------------------
# Признаки и целевая переменная (последний столбец)

X = data.iloc[:, :-1].values
y = data.iloc[:, -1].values

# ------------------------------------------------------------
# Кросс-валидация 5-fold с перемешиванием

kf = KFold(n_splits=5, shuffle=True, random_state=1)

# ------------------------------------------------------------
# Перебор числа деревьев от 1 до 50
best_n = None
for n in range(1, 51):
    rf = RandomForestRegressor(n_estimators=n, random_state=1)
    scores = []
    for train_idx, val_idx in kf.split(X):
        X_train, X_val = X[train_idx], X[val_idx]
        y_train, y_val = y[train_idx], y[val_idx]
        rf.fit(X_train, y_train)
        pred = rf.predict(X_val)
        scores.append(r2_score(y_val, pred))
    mean_r2 = np.mean(scores)
    if mean_r2 > 0.52:
        best_n = n
        break

print(f"Минимальное количество деревьев для R2 > 0.52: {best_n}")

```
