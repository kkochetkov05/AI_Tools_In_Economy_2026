# 13-statement-gbm

Размер обучающей выборки составляет 750 объектов, размер тестовой выборки — 3001 объект, каждый объект имеет 1776 признаков.

Построены графики значений log-loss на обучающей и тестовой выборках для каждого значения learning_rate из списка [1, 0.5, 0.3, 0.2, 0.1]. Для learning_rate = 0.2 минимальное значение log-loss на тестовой выборке достигается на 37 итерации.

При увеличении числа итераций качество на обучающей выборке 
продолжает улучшаться, а на тестовой после некоторого момента 
начинает ухудшаться. Это признак ПЕРЕОБУЧЕНИЯ (overfitting).

Минимальное значение log-loss на тестовой выборке при learning_rate = 0.2 равно 0.530519, номер итерации — 37.                                                                             
Log-loss случайного леса, обученного с количеством деревьев 37, на тестовой выборке равен 0.541094.

### 1. Загрузите выборку из файла gbm-data.csv

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import GradientBoostingClassifier, RandomForestClassifier
from sklearn.metrics import log_loss
import matplotlib.pyplot as plt


url = 'https://raw.githubusercontent.com/Sazankova/gbm/main/gbm-data.csv'
data = pd.read_csv(url)

data_np = data.values

y = data_np[:, 0]
X = data_np[:, 1:]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.8, random_state=241
)

print(f"Размер обучающей выборки: {X_train.shape}")
print(f"Размер тестовой выборки: {X_test.shape}")
```

### 2. Обучение GradientBoostingClassifier с разными learning_rate

```python
learning_rates = [1, 0.5, 0.3, 0.2, 0.1]


results = {}

for lr in learning_rates:
    print(f"\n=== Обучение с learning_rate = {lr} ===")
    
    
    clf = GradientBoostingClassifier(
        n_estimators=250,
        learning_rate=lr,
        verbose=True,
        random_state=241
    )
    clf.fit(X_train, y_train)
    
    
    train_loss = []
    test_loss = []
    
    
    for pred_train, pred_test in zip(
        clf.staged_decision_function(X_train),
        clf.staged_decision_function(X_test)
    ):
       
        proba_train = 1 / (1 + np.exp(-pred_train))
        proba_test = 1 / (1 + np.exp(-pred_test))
        
        
        train_loss.append(log_loss(y_train, proba_train))
        test_loss.append(log_loss(y_test, proba_test))
    
    
    min_test_loss = min(test_loss)
    best_iteration = test_loss.index(min_test_loss) + 1  # +1 т.к. итерации с 1
    
    results[lr] = {
        'train_loss': train_loss,
        'test_loss': test_loss,
        'min_test_loss': min_test_loss,
        'best_iteration': best_iteration
    }
    
    print(f"Лучшее качество на тесте: {min_test_loss:.6f} (итерация {best_iteration})")
    
    
    plt.figure(figsize=(10, 6))
    plt.plot(train_loss, 'r-', linewidth=2, label='Train loss')
    plt.plot(test_loss, 'g-', linewidth=2, label='Test loss')
    plt.axvline(x=best_iteration-1, color='b', linestyle='--', 
                label=f'Best iteration = {best_iteration}')
    plt.title(f'Log-loss при learning_rate = {lr}')
    plt.xlabel('Итерация')
    plt.ylabel('Log-loss')
    plt.legend()
    plt.grid(True)
    plt.show()
```

### 3. Анализ графика качества

```python
print("""
При увеличении числа итераций качество на обучающей выборке 
продолжает улучшаться, а на тестовой после некоторого момента 
начинает ухудшаться. Это признак ПЕРЕОБУЧЕНИЯ (overfitting).
""")
print("Ответ на вопрос 3: overfitting")
```

### 4. Результат для learning_rate = 0.2


```python
lr_target = 0.2
if lr_target in results:
    best_iter = results[lr_target]['best_iteration']
    min_loss = results[lr_target]['min_test_loss']
    
    print(f"\n=== Результаты для learning_rate = {lr_target} ===")
    print(f"Минимальное значение log-loss на тесте: {min_loss:.6f}")
    print(f"Номер итерации: {best_iter}")
    print(f"\nОтвет на вопрос 4: {min_loss:.6f} {best_iter}")
else:
    print(f"Данные для learning_rate = {lr_target} не найдены")
```

### 5. Сравнение со случайным лесом

```python
n_estimators_rf = results[0.2]['best_iteration']

print(f"\n=== Обучение RandomForestClassifier ===")
print(f"Количество деревьев: {n_estimators_rf}")

rf_clf = RandomForestClassifier(
    n_estimators=n_estimators_rf,
    random_state=241
)
rf_clf.fit(X_train, y_train)


rf_proba = rf_clf.predict_proba(X_test)
rf_pred = rf_proba[:, 1]

rf_log_loss = log_loss(y_test, rf_pred)

print(f"Log-loss случайного леса на тесте: {rf_log_loss:.6f}")
print(f"\nОтвет на вопрос 5: {rf_log_loss:.6f}")
```
