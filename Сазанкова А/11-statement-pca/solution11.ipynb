# 11-statement-pca

### 1. Загрузите данные close_prices.csv

```python
import pandas as pd
import numpy as np
from sklearn.decomposition import PCA

url_close = 'https://raw.githubusercontent.com/Sazankova/pca/main/close_prices.csv'
df = pd.read_csv(url_close)
X = df.drop('date', axis=1)
```

### 2. Обучите PCA с числом компонент 10. Сколько компонент хватит для 90% дисперсии?

```python
pca = PCA(n_components=10)
pca.fit(X)

explained_variance = pca.explained_variance_ratio_
cumsum = np.cumsum(explained_variance)

for i, cum in enumerate(cumsum):
    print(f"{i+1} компонент(ы): {cum:.4f}")

n_components_90 = np.argmax(cumsum >= 0.9) + 1
print(f"\nДля объяснения 90% дисперсии нужно {n_components_90} компонент(ы)")
```

### 3. Примените преобразование к исходным данным и возьмите первую компоненту

```python
X_pca = pca.transform(X)
first_component = X_pca[:, 0]
print("Первая компонента (первые 5 значений):")
print(first_component[:5])
```

### 4. Загрузите информацию об индексе Доу-Джонса

```python
url_djia = 'https://raw.githubusercontent.com/Sazankova/pca/main/djia_index.csv'
djia = pd.read_csv(url_djia)

print("Колонки в файле:", djia.columns.tolist())

if '^DJI' in djia.columns:
    correlation = np.corrcoef(first_component, djia['^DJI'])[0, 1]
elif 'DJIA' in djia.columns:
    correlation = np.corrcoef(first_component, djia['DJIA'])[0, 1]
else:
    correlation = np.corrcoef(first_component, djia.iloc[:, 0])[0, 1]

print(f"\nКорреляция Пирсона: {correlation:.2f}")
```

### 5. Какая компания имеет наибольший вес в первой компоненте?

```python
weights = pca.components_[0]
feature_names = X.columns
max_index = np.argmax(np.abs(weights))
company = feature_names[max_index]
weight_value = weights[max_index]

print(f"Компания с наибольшим весом: {company}")
print(f"Вес: {weight_value:.4f}")
```
