# 01-titanic-assignment

```python
import pandas as pd
import numpy as np
url = 'https://raw.githubusercontent.com/Sazankova/Titanic/main/titanic.csv'
df = pd.read_csv(url, index_col='PassengerId')
total = len(df)
```

# 1. Какое количество мужчин и женщин ехало на корабле? 

 ### В качестве ответа приведите два числа через пробел

```python
male_count = (df['Sex'] == 'male').sum()
female_count = (df['Sex'] == 'female').sum()
print(male_count, female_count)
```

# 2. Какой части пассажиров удалось выжить? 

###  Посчитайте долю выживших пассажиров. Ответ приведите в процентах (число в интервале от 0 до 100, знак процента не нужен), округлив до двух знаков.

```python
survived_pct = ((df['Survived'] == 1).sum() / total) * 100
print(f"{survived_pct:.2f}")
```

# Какую долю пассажиры первого класса составляли среди всех пассажиров?

### Ответ приведите в процентах (число в интервале от 0 до 100, знак процента не нужен), округлив до двух знаков.

```python
pclass1_pct = ((df['Pclass'] == 1).sum() / total) * 100
print(f"{pclass1_pct:.2f}")
```

# 4. Какого возраста были пассажиры?

### Посчитайте среднее и медиану возраста пассажиров. Посчитайте среднее и медиану возраста пассажиров. В качестве ответа приведите два числа через пробел.

```python
age_mean = df['Age'].mean()
age_median = df['Age'].median()
print(f"{age_mean:.2f} {age_median:.2f}")
```

# 5. Коррелируют ли число братьев/сестер с числом родителей/детей?

### Посчитайте корреляцию Пирсона между признаками SibSp и Parch.

```python
correlation = df['SibSp'].corr(df['Parch'])
print(f"{correlation:.2f}")
```

# 6. Какое самое популярное женское имя на корабле? 

### Извлеките изполного имени пассажира (колонка Name) его личное имя (FirstName). Это задание типичный пример того, с чем сталкивается специалист по анализу данных. Данные очень разнородные ишумные, но из них требуется извлечь необходимую информацию.Попробуйте вручную разобрать несколько значений столбца Name и выработать правило для извлечения имен, а также разделенияих на женские и мужские.

```python
female_names = {}

for name in df['Name']:
    if ' Mrs.' in name:
        if '(' in name:
            first_name = name.split('(')[1].split()[0].strip('()')
            female_names[first_name] = female_names.get(first_name, 0) + 1

    elif ' Miss.' in name:
        first_name = name.split('. ')[1].split()[0].strip('()')
        female_names[first_name] = female_names.get(first_name, 0) + 1

max_value = max(female_names.values())
max_keys = [key for key, value in female_names.items() if value == max_value]

print(max_keys, max_value)
```
