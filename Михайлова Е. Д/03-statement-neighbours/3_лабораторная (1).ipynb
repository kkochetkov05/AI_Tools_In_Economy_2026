1. Загрузите выборку Wine

```python
import numpy as np
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import KFold, cross_val_score
from sklearn.preprocessing import scale

df = pd.read_csv('winedata.txt', header=None)
```

2. Извлеките из данных признаки и классы. Класс записан в первом
столбце (три варианта), признаки — в столбцах со второго по последний. Более подробно о сути признаков можно прочитать по адресу https://archive.ics.uci.edu/ml/datasets/Wine (см. также файл
wine.names, приложенный к заданию)


```python
X = df.iloc[:, 1:].values
y = df.iloc[:, 0].values
```

3. Оценку качества необходимо провести методом кросс-валидации по
5 блокам (5-fold). Создайте генератор разбиений, который перемешивает выборку перед формированием блоков (shuffle=True). Для воспроизводимости результата, создавайте генератор KFold с фиксированным параметром random_state=42. В качестве меры качества используйте долю верных ответов (accuracy).


```python
kf = KFold(n_splits=5, shuffle=True, random_state=42)
```

4. Найдите точность классификации на кросс-валидации для метода
k ближайших соседей (sklearn.neighbors.KNeighborsClassifier), при
k от 1 до 50. При каком k получилось оптимальное качество? Чему
оно равно (число в интервале от 0 до 1)? Данные результаты и
будут ответами на вопросы 1 и 2.


```python
best_k_unscaled, best_acc_unscaled = None, 0.0
for k in range(1, 51):
    knn = KNeighborsClassifier(n_neighbors=k)
    scores = cross_val_score(knn, X, y, cv=kf, scoring='accuracy')
    acc = np.mean(scores)
    if acc > best_acc_unscaled:
        best_acc_unscaled = acc
        best_k_unscaled = k
print(f"{best_k_unscaled} {best_acc_unscaled:.2f}")
```

5. Произведите масштабирование признаков с помощью функции
sklearn.preprocessing.scale. Снова найдите оптимальное k на кроссвалидации.


```python
X_scaled = scale(X)
best_k_scaled, best_acc_scaled = None, 0.0
for k in range(1, 51):
    knn = KNeighborsClassifier(n_neighbors=k)
    scores = cross_val_score(knn, X_scaled, y, cv=kf, scoring='accuracy')
    acc = np.mean(scores)
    if acc > best_acc_scaled:
        best_acc_scaled = acc
        best_k_scaled = k
print(f"{best_k_scaled} {best_acc_scaled:.2f}")
with open('result3.txt','w', encoding='utf-8') as f:
  f.write(str(round(best_acc_unscaled, 2)) + '\n')
  f.write(str(best_k_unscaled) + '\n')
  f.write(str(round(best_acc_scaled, 2)) + '\n')
  f.write(str(best_k_scaled))
```
