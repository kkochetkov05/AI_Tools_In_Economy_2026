6.  Какое самое популярное женское имя на корабле? Извлеките из
полного имени пассажира (колонка Name) его личное имя (First
Name). Это задание — типичный пример того, с чем сталкивается специалист по анализу данных. Данные очень разнородные и
шумные, но из них требуется извлечь необходимую информацию.
Попробуйте вручную разобрать несколько значений столбца Name
и выработать правило для извлечения имен, а также разделения
их на женские и мужские.


```python
import re
def extract_female_first_name(name):
    parts = name.split(',')
    if len(parts) < 2:
        return None
    rest = parts[1].strip()
    female_titles = re.compile(r'\b(Mrs?|Miss|Ms|Lady|Countess|Mlle|Mme|the Countess)\b', re.IGNORECASE)
    match = female_titles.search(rest)
    if not match:
        return None
    title = match.group(0)
    after_title = rest[match.end():].strip()

    #если есть скобки то девичья фамилия или имя
    bracket_match = re.search(r'\(([^)]+)\)', after_title)
    if bracket_match:
        inside = bracket_match.group(1)
        inside = re.sub(r'["\']', '', inside)
        first = inside.split()[0].rstrip('.')
        return first
    #иначе берем первое слово после титула
    after_title = re.sub(r'^[,.\s]+', '', after_title)
    if not after_title:
        return None
    first = after_title.split()[0].rstrip('.')
    first = re.sub(r'["\']', '', first)
    if len(first) < 2:
        return None
    return first

data['FirstName'] = data['Name'].apply(extract_female_first_name)

female_df = data[data['Sex'] == 'female'].copy()
female_firstnames = female_df['FirstName'].dropna()

most_common = female_firstnames.value_counts().idxmax()
freq = female_firstnames.value_counts()
max_count = freq.max()
top_names = freq[freq == max_count].index.tolist()
result6 = f"{' '.join(top_names)} "
print(f"{result6}")
with open('result1.txt','w',encoding='utf-8') as f:
  f.write(result1)
  f.write(f"\n{result2}"\n{result3}\n{result4}\n{correlation}\n{result6}")
```
