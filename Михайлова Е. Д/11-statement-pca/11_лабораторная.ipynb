1. Загрузите данные close_prices.csv. В этом файле приведены цены
акций 30 компаний на закрытии торгов за каждый день периода.

```python
import pandas as pd
import numpy as np
from sklearn.decomposition import PCA
from scipy.stats import pearsonr

prices = pd.read_csv('close_prices.csv', index_col=0)
X = prices.values
```

2. На загруженных данных обучите преобразование PCA с числом
компоненты равным 10. Скольких компонент хватит, чтобы объяснить 90% дисперсии?

```python
pca = PCA(n_components=10)
pca.fit(X)

cumsum = np.cumsum(pca.explained_variance_ratio_)
n_90 = np.searchsorted(cumsum, 0.9) + 1
print(f"{n_90}")
```

3. Примените построенное преобразование к исходным данным и возьмите значения первой компоненты

```python
first_component = pca.transform(X)[:, 0]
```

4. Загрузите информацию об индексе Доу-Джонса из файла djia_prices.csv.
Чему равна корреляция Пирсона между первой компонентой и индексом Доу-Джонса?

```python
djia = pd.read_csv('djia_index.csv')
merged = pd.merge(prices.reset_index(), djia, on='date', how='inner')
first_comp_aligned = first_component[:len(merged)]
corr, _ = pearsonr(first_comp_aligned, merged['^DJI'].values)
print(f"{corr:.2f}")
```

5. Какая компания имеет наибольший вес в первой компоненте?

```python
loadings = pca.components_[0]
max_idx = np.argmax(np.abs(loadings))
company = prices.columns[max_idx]
print(f"{company}")
with open('result11.txt','w',encoding='utf-8') as f:
    f.write(str(n_90))
    f.write(f"\n{corr:.2f}")
    f.write(f"\n{company}")
```
