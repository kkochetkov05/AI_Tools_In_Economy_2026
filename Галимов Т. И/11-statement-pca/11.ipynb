### Полный код для задания

```python
#Общий код для задания
import pandas as pd
import numpy as np
from sklearn.decomposition import PCA

data = pd.read_csv('close_prices.csv',index_col = 'date')
clf = PCA(n_components = 10)
clf.fit(data)

#print(clf.explained_variance_ratio_.sum()) #4 компоненты для 90%
with open ('2.txt','w') as file:
    file.write(str(4))

clf.transform(data)

djia = pd.read_csv('djia_index.csv',index_col = 'date')
comp = []
for i in range(len(data)):
    comp.append(clf.transform(data)[i][0])
Pirson = np.corrcoef(comp,djia.iloc[:,0])[0,1]
with open('4.txt','w') as file:
    file.write(str(round(Pirson,2)))
w_max = np.argmax(clf.components_[0])
with open('5.txt','w') as file:
    file.write(str(data.columns[w_max - 1]))
```
