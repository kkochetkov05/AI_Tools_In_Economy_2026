### 1. Загрузите выборку из файла svm-data.csv. Целевая переменная находится в первом столбце, признаки — во втором и третьем.

```python
import pandas as pd
import numpy as np
from sklearn.svm import SVC

data = pd.read_csv('svm-data.csv', header = None)
```

### 2. Обучите классификатор с линейным ядром, параметром C=100000 и random_state=241.

```python
X = data.iloc[:, 1:].values
y = data.iloc[:, 0].values

clf = SVC(kernel = 'linear', C = 100000, random_state=241)
clf.fit(X, y)
```

### 3. Найдите номера объектов, которые являются опорными. Нумерация объектов в файле начинается с 1.

```python
support_idx = clf.support_
support_idx = sorted(support_idx + 1)

ans = ','.join(map(str, support_idx))
print(ans)
```

```python
#Общий код для задания
import pandas as pd
import numpy as np
from sklearn.svm import SVC

data = pd.read_csv('svm-data.csv', header = None)

X = data.iloc[:, 1:].values
y = data.iloc[:, 0].values

clf = SVC(kernel = 'linear', C = 100000, random_state=241)
clf.fit(X,y)

support_idx = clf.support_
support_idx = sorted(support_idx + 1)

ans = ','.join(map(str,support_idx))

with open('ans.txt','w') as file:
    file.write(ans)
```
