### 6. Какое самое популярное женское имя на корабле?
Извлеките из
полного имени пассажира (колонка Name) его личное имя (First
Name).

```python
women = data[data['Sex'] == 'female'].copy()
def popular_name(name):
    if 'Miss.' in name:
        part = name.split('Miss.')[1].strip()
        return part.split()[0]
    elif 'Mrs.' in name:
        if '(' in name:
            part = name.split('(')[1].split(')')[0]
            return part.split()[0]
        else:
            part = name.split('Mrs.')[1].strip()
            return part.split()[0]
    return None

women['first_name'] = women['Name'].apply(popular_name)

print(str(women['first_name'].value_counts().index[0]))
```
