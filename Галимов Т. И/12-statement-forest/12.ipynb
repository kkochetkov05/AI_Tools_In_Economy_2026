### 1. Загрузите данные из файла abalone.csv. Это датасет, в котором требуется предсказать возраст ракушки (число колец) по физическим измерениям.

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import KFold
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score

data = pd.read_csv('abalone.csv')
```

### 2. Преобразуйте признак Sex в числовой: значение F должно перейти в -1, I — в 0, M — в 1.

```python
data['Sex'] = data['Sex'].map(lambda x: 1 if x == 'M' else (-1 if x == 'F' else 0))
```

### 3. Разделите содержимое файлов на признаки и целевую переменную. В последнем столбце записана целевая переменная, в остальных — признаки.

```python
X = data.drop('Rings', axis=1).values
y = data['Rings'].values
```

### 4. Обучите случайный лес (sklearn.ensemble.RandomForestRegressor) с различным числом деревьев: от 1 до 50 (random_state=1). Для каждого из вариантов оцените качество работы полученного леса на кросс-валидации по 5 блокам. Используйте параметры random_state=1 и shuffle=True при создании генератора кросс-валидации sklearn.model_selection.KFold. В качестве меры качества воспользуйтесь коэффициентом детерминации (sklearn.metrics.r2_score).

```python
kf = KFold(n_splits=5, random_state=1, shuffle=True)

for i in range(50):
    scores = []
    for train_idx, test_idx in kf.split(X):
        X_train = X[train_idx]
        y_train = y[train_idx]
        X_test = X[test_idx]
        y_test = y[test_idx]
        clf = RandomForestRegressor(n_estimators=i+1, random_state=1)
        clf.fit(X_train, y_train)
        prediction = clf.predict(X_test)
        score = r2_score(y_test, prediction)
        scores.append(score)
    mean_score = np.mean(scores)
    if mean_score >= 0.52:
        best_n = i + 1
        break
```

### 5. Определите, при каком минимальном количестве деревьев случайный лес показывает качество на кросс-валидации выше 0.52. Это количество и будет ответом на задание.

```python
print(best_n)
```

```python
#Общий код для задания
import pandas as pd
import numpy as np
from sklearn.model_selection import KFold
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score

data = pd.read_csv('abalone.csv')
data['Sex'] = data['Sex'].map(lambda x: 1 if x == 'M' else (-1 if x == 'F' else 0))

X = data.drop('Rings',axis = 1).values
y = data['Rings'].values
kf = KFold(n_splits = 5, random_state = 1, shuffle = True)

for i in range(50):
    scores = []
    for train_idx,test_idx in kf.split(X):
        X_train = X[train_idx]
        y_train = y[train_idx]
        X_test = X[test_idx]
        y_test = y[test_idx]
        clf = RandomForestRegressor(n_estimators = i+1,random_state = 1)
        clf.fit(X_train,y_train)
        prediction = clf.predict(X_test)
        score = r2_score(y_test,prediction)
        scores.append(score)
    mean_score = np.mean(scores)
    print(mean_score)
    if mean_score >= 0.52:
        best_n = i + 1
        break

with open ('ans.txt','w') as file:
    file.write(str(best_n))
```
