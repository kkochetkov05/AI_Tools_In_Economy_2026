```python
import pandas as pd
import re
```

```python
data = pd.read_csv('titanic.csv', index_col='PassengerId')
```

1. Какое количество мужчин и женщин ехало на корабле?

```python
print(data['Sex'].value_counts()['male'], data['Sex'].value_counts()['female'])
```

2. Какой части пассажиров удалось выжить?

```python
print(round(100 * data['Survived'].value_counts(1)[1], 2))
```

3. Какую долю пассажиры первого класса составляли среди всех пассажиров?

```python
print(round(100 * data['Pclass'].value_counts(1)[1], 2))
```

4. Какого возраста были пассажиры? 

```python
print(data['Age'].mean(), data['Age'].median())
```

5. Коррелируют ли число братьев/сестер с числом родителей/детей?

```python
print(data['SibSp'].corr(data['Parch']))
```

6. Какое самое популярное женское имя на корабле? 

```python
def extract_first_name(name):
    match = re.search(r'\((.*?)\)', name)
    if match:
        return match.group(1).split()[0]
    
    parts = name.split('. ')
    if len(parts) > 1:
        return parts[1].split()[0]
    
    return None

women = data[data['Sex'] == 'female']
women['FirstName'] = women['Name'].apply(extract_first_name)

top_name = women['FirstName'].value_counts().idxmax()

print(top_name)
```
