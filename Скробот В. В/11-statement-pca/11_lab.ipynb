1. Загрузите данные close_prices.csv. В этом файле приведены цены акций 30 компаний на закрытие торгов за каждый день периода.

```python
import pandas as pd

dataX = pd.read_csv('close_prices.csv')
dataX = dataX.set_index('date')
```

2. На загруженных данных обучите преобразование PCA с числом компоненты равным 10. Скольких компонент хватит, чтобы объяснить 90% дисперсии?

```python
from sklearn.decomposition import PCA
dec = PCA(n_components=10)
dataX_tr = dec.fit_transform(dataX)



sum_var = 0
for i in range(len(dec.explained_variance_ratio_)):
    sum_var = dec.explained_variance_ratio_[i] + sum_var
    if sum_var >= 0.9:
        break
print(i+1)
```
