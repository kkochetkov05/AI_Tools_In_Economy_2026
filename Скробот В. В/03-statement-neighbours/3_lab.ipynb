1. Загрузите выборку Wine по адресу https://archive.ics.uci.edu/ml/machine-learning-databases/wine/wine.data

```python
import pandas as pd
import numpy as np
data = pd.read_csv('wine.data', header=None)

data.columns = ['class', 'Alcohol', 'Malicacid', 'Ash', 'Alcalinity_of_ash', 'Magnesium', 
               'Total_phenols', 'Flavanoids', 'Nonflavanoid_phenols', 'Proanthocyanins', 'Color_intensity', 
               'Hue', '0D280_0D315_of_diluted_wines', 'Proline']
```

2. Извлеките из данных признаки и классы. Класс записан в первом столбце (три варианта), признаки — в столбцах со второго по последний.

```python
data.columns = ['class', 'Alcohol', 'Malicacid', 'Ash', 'Alcalinity_of_ash', 'Magnesium', 
               'Total_phenols', 'Flavanoids', 'Nonflavanoid_phenols', 'Proanthocyanins', 'Color_intensity', 
               'Hue', '0D280_0D315_of_diluted_wines', 'Proline']
X = data.drop(columns=['class'])
y = data['class']
```

3. Оценку качества необходимо провести методом кросс-валидации по 5 блокам (5-fold). Создайте генератор разбиений, который перемешивает выборку перед формированием блоков (shuffle=True). Для
воспроизводимости результата, создавайте генератор KFold с фиксированным параметром random_state=42. В качестве меры качества используйте долю верных ответов (accuracy).

```python
from sklearn.model_selection import KFold
from sklearn.model_selection import cross_val_score
kf = KFold(n_splits=5, shuffle=True, random_state=42)
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

4. Найдите точность классификации на кросс-валидации для метода к ближайших соседей (sklearn.neighbors.KNeighborsClassifier), при
к от 1 до 50. При каком к получилось оптимальное качество? Чему оно равно (число в интервале от 0 до 1)? Данные результаты и будут ответами на вопросы 1 и 2.

```python
best_k1 = None
best_score1 = -np.inf
best_k2 = None
best_score2 = -np.inf
for k in range(1,51):
    KNC = KNeighborsClassifier(n_neighbors=k)
    scores = cross_val_score(KNC, X, y, cv=kf, scoring='accuracy')
    mean_score = scores.mean()
    if mean_score > best_score1:
        best_score1 = mean_score
        best_k1 = k
print(best_k1, best_score1)
```

5. Произведите масштабирование признаков с помощью функции sklearn.preprocessing.scale. Снова найдите оптимальное к на кросс-валидации.

```python
from sklearn.preprocessing import scale
X_scaled = scale(X)
for k in range(1,51):
    KNC = KNeighborsClassifier(n_neighbors=k)
    scores = cross_val_score(KNC, X_scaled, y, cv=kf, scoring='accuracy')
    mean_score = scores.mean()
    if mean_score > best_score2:
        best_score2 = mean_score
        best_k2 = k
```

6. Какое значение к получилось оптимальным после приведения признаков к одному масштабу? Как изменилось значение качества? Приведите ответы на вопросы 3 и 4.

```python
print(best_k2, best_score2)
```
