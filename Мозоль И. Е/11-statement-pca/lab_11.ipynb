### **2. На загруженных данных обучите преобразование PCA с числом компоненты равным 10.**

Сколько компонент хватит, чтобы объяснить 90% дисперсии?

```python
close_prices = pd.read_csv('close_prices.csv')
X = close_prices.drop('date', axis=1).values if 'date' in close_prices.columns else close_prices.values

pca = PCA(n_components=10)
pca.fit(X)

cumsum = np.cumsum(pca.explained_variance_ratio_)
n_90 = np.argmax(cumsum >= 0.9) + 1

print(f"Компонент для 90% дисперсии: {n_90}")

with open('answer1.txt', 'w') as f:
    f.write(str(n_90))
```

### **3. Примените построенное преобразование к исходным данным и возьмите значения первой компоненты.**

```python
import pandas as pd
from sklearn.decomposition import PCA

close_prices = pd.read_csv('close_prices.csv')
X = close_prices.drop('date', axis=1).values if 'date' in close_prices.columns else close_prices.values

pca = PCA(n_components=10)
pca.fit(X)
first_component = pca.transform(X)[:, 0]

print(f"Первые 5 значений: {first_component[:5]}")
```

### **4. Загрузите информацию об индексе Доу-Джонса из файла djia_prices.csv.**

Чему равна корреляция Пирсона между первой компонентой и индексом Доу-Джонса?

```python
djia = pd.read_csv('djia_index.csv')
djia_values = djia['DJIA'].values if 'DJIA' in djia.columns else djia.iloc[:, -1].values

correlation = np.corrcoef(first_component, djia_values)[0, 1]
answer = f"{correlation:.2f}"

print(f"Корреляция Пирсона: {answer}")

with open('answer2.txt', 'w') as f:
    f.write(answer)
```
