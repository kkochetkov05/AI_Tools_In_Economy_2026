```python
import pandas as pd
import numpy as np
import sklearn as sk
```

1. Загрузите обучающую и тестовую выборки из файлов perceptrontrain.csv и perceptron-test.csv. Целевая переменная записана в первом столбце, признаки — во втором и третьем.

```python
train_data = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/ML_intro/05/perceptron-train.csv', header=0, names=['target', 'feauture1', 'feauture2'])
test_data = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/ML_intro/05/perceptron-test.csv', header=0, names=['target', 'feauture1', 'feauture2'])
```

```python
train_y = train_data['target'].to_numpy()
train_X = train_data.drop(columns=['target']).to_numpy()


test_y = test_data['target'].to_numpy()
test_X = test_data.drop(columns=['target']).to_numpy()
```

2. Обучите персептрон со стандартными параметрами и random_state=241.

```python
clf = sk.linear_model.Perceptron(random_state=241)
clf.fit(train_X, train_y)
predicted_y = clf.predict(test_X)
```

3. Подсчитайте качество (долю правильно классифицированных объектов, accuracy) полученного классификатора на тестовой выборке.

```python
score = sk.metrics.accuracy_score(test_y, predicted_y)
print(round(score, 3))
```

4. Нормализуйте обучающую и тестовую выборку с помощью класса
StandardScaler.


```python
scaler = sk.preprocessing.StandardScaler()

train_X_scaled = scaler.fit_transform(train_X)
test_X_scaled = scaler.transform(test_X)
```

5. Обучите персептрон на новых выборках. Найдите долю правильных ответов на тестовой выборке.

```python
clf = sk.linear_model.Perceptron(random_state=241)
clf.fit(train_X_scaled, train_y)
predicted_y_scaled = clf.predict(test_X_scaled)
score_scaled = sk.metrics.accuracy_score(test_y, predicted_y_scaled)
print(round(score_scaled, 3))
```

6. Найдите разность между качеством на тестовой выборке после нормализации и качеством до нее. Это число и будет ответом на задание.

```python
print(round(score_scaled - score, 3))
```
