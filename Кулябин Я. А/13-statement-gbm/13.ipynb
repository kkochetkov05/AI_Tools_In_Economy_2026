```python
import numpy as np
import pandas as pd
from sklearn.ensemble import GradientBoostingClassifier, RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import log_loss
import matplotlib.pyplot as plt
```

1. Загрузите выборку из файла gbm-data.csv с помощью pandas и пре
образуйте ее в массив numpy (параметр values у датафрейма). В
3
первой колонке файла с данными записано, была или нет реакция.
Все остальные колонки (d1- d1776) содержат различные характери
стики молекулы, такие как размер, форма и т.д. Разбейте выборку
на обучающую и тестовую, используя функцию train_test_split с
параметрами test_size = 0.8 и random_state = 241.

```python
data = pd.read_csv('gbm-data.csv')
data.head()
```

```python
y = data.iloc[:, 0].values
X = data.iloc[:, 1:].values

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.8, random_state=241
)
```

2. ОбучитеGradientBoostingClassifier с параметрами n_estimators=250,
verbose=True, random_state=241 и для каждого значения learning_rate
из списка [1, 0.5, 0.3, 0.2, 0.1] проделайте следующее:
• Используйте метод staged_decision_function для предсказа
ния качества на обучающей и тестовой выборке на каждой
итерации.
• Преобразуйте полученное предсказание по формуле 1
1+e−y_pred
,
где y_pred предсказанное значение.
• Вычислите и постройте график значений log-loss на обучаю
щей и тестовой выборках, а также найдите минимальное зна
чение метрики и номер итерации, на которой оно достигается.

```python
learning_rates = [1, 0.5, 0.3, 0.2, 0.1]
results = {}

for lr in learning_rates:
    print(f"\n--- learning_rate = {lr} ---")
    
    # Создаем и обучаем модель
    gbc = GradientBoostingClassifier(
        n_estimators=250,
        learning_rate=lr,
        random_state=241,
        verbose=True
    )
    
    gbc.fit(X_train, y_train)
    
    # Получаем предсказания на каждой итерации
    train_staged_pred = gbc.staged_decision_function(X_train)
    test_staged_pred = gbc.staged_decision_function(X_test)
    
    # Преобразуем отступы в вероятности через сигмоиду
    # вероятность = 1 / (1 + exp(-decision_function))
    train_loss = []
    test_loss = []
    
    for train_pred, test_pred in zip(train_staged_pred, test_staged_pred):
        # Преобразуем в вероятности
        train_proba = 1.0 / (1.0 + np.exp(-train_pred))
        test_proba = 1.0 / (1.0 + np.exp(-test_pred))
        
        # Вычисляем log-loss
        train_loss.append(log_loss(y_train, train_proba))
        test_loss.append(log_loss(y_test, test_proba))
    
    # Находим минимальное значение и номер итерации
    train_loss = np.array(train_loss)
    test_loss = np.array(test_loss)
    
    min_test_loss = np.min(test_loss)
    min_test_iter = np.argmin(test_loss) + 1  # +1 для нумерации с 1
    
    min_train_loss = np.min(train_loss)
    min_train_iter = np.argmin(train_loss) + 1
    
    results[lr] = {
        'train_loss': train_loss,
        'test_loss': test_loss,
        'min_test_loss': min_test_loss,
        'min_test_iter': min_test_iter,
        'min_train_loss': min_train_loss,
        'min_train_iter': min_train_iter
    }
    
    print(f"\nРезультаты для learning_rate={lr}:")
    print(f"  Минимальный train loss: {min_train_loss:.6f} (итерация {min_train_iter})")
    print(f"  Минимальный test loss:  {min_test_loss:.6f} (итерация {min_test_iter})")
    
    # Строим график
    plt.figure(figsize=(12, 6))
    plt.plot(train_loss, 'b', linewidth=2, label='Train')
    plt.plot(test_loss, 'r', linewidth=2, label='Test')
    plt.xlabel('Номер итерации')
    plt.ylabel('Log-loss')
    plt.title(f'Зависимость качества от итерации (learning_rate={lr})')
    plt.legend()
    plt.grid(True, alpha=0.3)
    
    # Отмечаем минимум на тесте
    plt.axvline(x=min_test_iter-1, color='g', linestyle='--', alpha=0.7)
    plt.plot(min_test_iter-1, min_test_loss, 'go', markersize=8)
    plt.annotate(f'min: {min_test_loss:.4f}',
                 xy=(min_test_iter-1, min_test_loss),
                 xytext=(min_test_iter-1 + 20, min_test_loss + 0.02),
                 arrowprops=dict(arrowstyle='->', color='green'))
    
    plt.tight_layout()
    plt.show()
```

3. Как можно охарактеризовать график качества на тестовой выбор
ке, начиная с некоторой итерации: переобучение (overfitting) или
недообучение (underfitting)? В ответе укажите одно из слов overfitting
либо underfitting.

```python
# Для learning_rate=0.2 анализируем поведение
lr_for_analysis = 0.2
test_loss = results[lr_for_analysis]['test_loss']
min_iter = results[lr_for_analysis]['min_test_iter']

# Смотрим, как ведет себя ошибка после минимума
after_min = test_loss[min_iter:]
if len(after_min) > 10:
    last_10_mean = np.mean(after_min[-10:])
    min_value = results[lr_for_analysis]['min_test_loss']
    
    if last_10_mean > min_value * 1.05:  # ошибка выросла более чем на 5%
        behavior = "overfitting"
    else:
        behavior = "underfitting"
else:
    behavior = "underfitting"
    
print(f"  Поведение после минимума: {behavior}")
```

4. Приведите минимальное значение log-loss на тестовой выборке и
номер итерации, на котором оно достигается, при learning_rate =
0.2.

```python
lr_02 = results[0.2]
print(f"Минимальное значение log-loss на тестовой выборке: {lr_02['min_test_loss']:.6f}")
print(f"Номер итерации, на котором оно достигается: {lr_02['min_test_iter']}")
```

5. На этих же данных обучите RandomForestClassifier с количеством
деревьев, равным количеству итераций, на котором достигается
наилучшее качество у градиентного бустинга из предыдущего пунк
та, random_state=241 и остальными параметрами по умолчанию.
Какое значение log-loss на тесте получается у этого случайного ле
са? (Не забывайте, что предсказания нужно получать с помощью
функции predict_proba. В данном случае брать сигмоиду от оценки
вероятности класса не нужно)

```python
# Обучаем случайный лес с количеством деревьев = оптимальной итерации GB
n_trees = lr_02['min_test_iter']
print(f"Количество деревьев в случайном лесе: {n_trees}")

rf = RandomForestClassifier(n_estimators=n_trees, random_state=241)
rf.fit(X_train, y_train)

# Предсказания вероятностей
rf_pred_proba = rf.predict_proba(X_test)

# Вычисляем log-loss
rf_log_loss = log_loss(y_test, rf_pred_proba)
print(f"Log-loss случайного леса на тесте: {rf_log_loss:.6f}")
print(f"Log-loss градиентного бустинга на тесте: {lr_02['min_test_loss']:.6f}")

# Сравнение
if rf_log_loss < lr_02['min_test_loss']:
    print("Случайный лес показал лучшее качество")
else:
    print("Градиентный бустинг показал лучшее качество")
```
