```python
df = pd.read_csv('data-logistic.csv', header = None)
df.head()
```

3. Реализуйте градиентный спуск для обычнойиL2-регуляризованной
(с коэффициентом регуляризации 10) логистической регрессии. Ис
пользуйте длину шага k=0.1. В качестве начального приближения
используйте вектор (0, 0)

```python
y = df[0].values
X = df[[1, 2]].values

def sigmoid(z):
    """Сигмоидная функция"""
    return 1 / (1 + np.exp(-z))

def logistic_gradient_descent(X, y, C=0, learning_rate=0.1, eps=1e-5, w_init = [], max_iter=10000):
    n_samples = X.shape[0]
    n_features = X.shape[1]
    
    # Начальное приближение (0, 0)
    if w_init:
        w = np.array(w_init)
    else:
        w = np.zeros(n_features)
    w_prev = w.copy()
    
    for iteration in range(max_iter):
        # Вычисляем линейную комбинацию
        linear = X @ w
        
        # Вычисляем сигмоиду
        sigma = sigmoid(-y * linear)
        
        # Вычисляем градиент
        grad = (1/n_samples) * (y[:, np.newaxis] * X * (1 - sigma)[:, np.newaxis]).sum(axis=0) - C * w
        
        # Обновляем веса
        w = w + learning_rate * grad
        
        # Проверяем сходимость
        if np.linalg.norm(w - w_prev) < eps:
            print(f"Сходимость достигнута на итерации {iteration + 1}")
            break
        
        w_prev = w.copy()
    
    if iteration == max_iter - 1:
        print(f"Достигнуто максимальное число итераций ({max_iter})")
    
    return w
```

4. Запустите градиентный спуск и доведите до сходимости (евклидово
расстояние между векторами весов на соседних итерациях долж
но быть не больше 1e-5). Рекомендуется ограничить сверху число
итераций десятью тысячами

```python
w_no_reg = logistic_gradient_descent(X, y, C=0, learning_rate=0.1)
print(f"Веса без регуляризации: w1 = {w_no_reg[0]:.6f}, w2 = {w_no_reg[1]:.6f}")

w_reg = logistic_gradient_descent(X, y, C=10, learning_rate=0.1)
print(f"Веса с регуляризацией: w1 = {w_reg[0]:.6f}, w2 = {w_reg[1]:.6f}")
```

5. Какое значение принимает AUC-ROC на обучении без регуляри
зации и при ее использовании? Эти величины будут ответом на
задание. В качестве ответа приведите два числа через пробел. Об
ратите внимание, что на вход функции roc_auc_score нужно пода
вать оценки вероятностей, подсчитанные обученным алгоритмом.
Для этого воспользуйтесь сигмоидной функцией: a(x) = 1/(1 +
exp(−w1x1 −w2x2)).

```python
# Оценки вероятностей: a(x) = 1/(1 + exp(-w1*x1 - w2*x2))
def predict_proba(X, w):
    """Предсказание вероятности принадлежности к классу 1"""
    linear = X @ w
    return sigmoid(linear)

# Предсказания без регуляризации
proba_no_reg = predict_proba(X, w_no_reg)
auc_no_reg = roc_auc_score(y, proba_no_reg)

# Предсказания с регуляризацией
proba_reg = predict_proba(X, w_reg)
auc_reg = roc_auc_score(y, proba_reg)
print(round(auc_no_reg, 3), round(auc_reg, 3))
```
