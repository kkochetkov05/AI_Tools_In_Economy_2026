6. Какое самое популярное женское имя на корабле? Извлеките из
полного имени пассажира (колонка Name) его личное имя (First
Name). Это задание типичный пример того, с чем сталкивает
ся специалист по анализу данных. Данные очень разнородные и
шумные, но из них требуется извлечь необходимую информацию.
Попробуйте вручную разобрать несколько значений столбца Name
и выработать правило для извлечения имен, а также разделения
их на женские и мужские.

```python
#6
def get_woman_first_name(row):
    """Извлекает только имя (без фамилии) для женщины"""
    name = row['Name']
    sex = row['Sex']
    
    if pd.isna(name) or sex != 'female':
        return None
    
    # Пытаемся найти имя в скобках
    match = re.search(r'\((?:.*?)([A-Za-z]+)(?:\s+[A-Za-z]+)*\s+[A-Za-z]+\)', name)
    if match:
        # Берем первое слово из скобок
        bracket_content = re.search(r'\((.*?)\)', name).group(1)
        first_name = bracket_content.split()[0]
        return first_name
    
    # Если нет скобок, берем первое слово после точки
    match = re.search(r'\.\s*([A-Za-z]+)', name)
    if match:
        return match.group(1)
    
    return None

data['Woman_Real_Name'] = data.apply(get_woman_first_name, axis=1)
first_name = data[data['Sex'] == 'female'][['Woman_Real_Name']]
#print(first_name)
name_counts = data['Woman_Real_Name'].value_counts()
```
