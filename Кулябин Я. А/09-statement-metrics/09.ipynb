• для логистической регрессии вероятность положительного

• дляSVM отступотразделяющейповерхности(колонкаscore_svm),
2
• для метрического алгоритма взвешенная сумма классов со

• для решающего дерева доля положительных объектов в ли

Загрузите этот файл

```python
df_scores = pd.read_csv('scores.csv')
df_scores.head()
```

5. Посчитайте площадь под ROC-кривой для каждого классифика
тора. Какой классификатор имеет наибольшее значение метрики
AUC-ROC (укажите название столбца с ответами этого классифи
катора)? Воспользуйтесь функцией sklearn.metrics.roc_auc_score

```python
y_true = df_scores['true']
classifiers = ['score_logreg', 'score_svm', 'score_knn', 'score_tree']
auc_scores = {}

for clf in classifiers:
    auc = roc_auc_score(y_true, df_scores[clf])
    auc_scores[clf] = auc
    print(f"{clf}: {auc:.4f}")

best_auc_clf = max(auc_scores, key=auc_scores.get)
best_auc_value = auc_scores[best_auc_clf]

print(f"\nКлассификатор с наибольшим AUC-ROC: {best_auc_clf}")
print(f"Значение AUC-ROC: {best_auc_value:.3f}")
```

6. Какой классификатор достигает наибольшей точности (Precision)
при полноте (Recall) не менее 70% (укажите название столбца с от
ветами этого классификатора)? Какое значение точности при этом
получается?

```python
results = {}
for clf in classifiers:
    precision_vals, recall_vals, thresholds = precision_recall_curve(y_true, df_scores[clf])
    max_precision = 0
    for p, r in zip(precision_vals, recall_vals):
        if r >= 0.7 and p > max_precision:
            max_precision = p
    
    results[clf] = max_precision

best_precision_clf = max(results, key=results.get)
best_precision_value = results[best_precision_clf]

print(f"\nКлассификатор с наибольшей точностью при полноте >= 0.7: {best_precision_clf}")
print(f"Значение точности: {best_precision_value:.3f}")
```
