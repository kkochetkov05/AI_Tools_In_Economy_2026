```python
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import KFold, cross_val_score
from sklearn.metrics import r2_score
```

1. Загрузите данные из файла abalone.csv. Это датасет, в котором тре
буется предсказать возраст ракушки (число колец) по физическим
измерениям.

```python
data = pd.read_csv('abalone.csv')
data.head()
```

2. Преобразуйте признак Sex в числовой: значение F должно перейти
в-1, I в 0, M в 1. Если вы используете Pandas, то подойдет следующий код: data[’Sex’] = data[’Sex’].map(lambda x: 1 if x ==
’M’ else (-1 if x == ’F’ else 0))

```python
data['Sex'] = data['Sex'].map(lambda x: 1 if x == 'M' else (-1 if x == 'F' else 0))
```

3. Разделите содержимое файлов на признаки и целевую переменную.
В последнем столбце записана целевая переменная, в остальных 
признаки.

```python
X = data.iloc[:, :-1].values
y = data.iloc[:, -1].values
```

4. Обучите случайный лес (sklearn.ensemble.RandomForestRegressor)
с различным числом деревьев: от 1 до 50 (random_state=1). Для
каждого из вариантов оцените качество работы полученного леса
на кросс-валидации по 5 блокам. Используйте параметры
"random_state=1"и "shuffle=True"при создании генератора кросс
валидации sklearn.cross_validation.KFold. В качестве меры качества
воспользуйтесь коэффициентом детерминации (sklearn.metrics.r2_score).

```python
# Параметры кросс-валидации
n_folds = 5
kfold = KFold(n_splits=n_folds, shuffle=True, random_state=1)

# Перебираем количество деревьев от 1 до 50
n_trees_range = range(1, 51)
r2_scores = []

for n_trees in n_trees_range:
    # Создаем случайный лес с заданным количеством деревьев
    rf = RandomForestRegressor(n_estimators=n_trees, random_state=1)
    
    # Вычисляем кросс-валидацию
    # Используем scoring='r2' для коэффициента детерминации
    scores = cross_val_score(rf, X, y, cv=kfold, scoring='r2')
    mean_score = scores.mean()
    r2_scores.append(mean_score)
    
    if n_trees % 5 == 0 or n_trees == 1:
        print(f"{n_trees:7d} | {mean_score:.6f}")
```

5. Определите, при каком минимальном количестве деревьев случай
ный лес показывает качество на кросс-валидации выше 0.52. Это
количество и будет ответом на задание.

```python
min_trees = None
for n_trees, score in zip(n_trees_range, r2_scores):
    if score > 0.52:
        min_trees = n_trees
        break

print(f"Первое количество деревьев с R2 > 0.52: {min_trees}")
print(f"Соответствующее значение R2: {r2_scores[min_trees-1]:.6f}")
```
