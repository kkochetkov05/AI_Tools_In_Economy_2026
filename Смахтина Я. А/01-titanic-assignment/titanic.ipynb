```python
import pandas as pd
data = pd.read_csv("titanic.csv", index_col="PassengerId")
```

1. Какое количество мужчин и женщин ехало на корабле? В качестве
ответа приведите два числа через пробел.


```python
sex_counts = data["Sex"].value_counts()
print(sex_counts.get('male', 0), sex_counts.get('female', 0))
```

2. Какой части пассажиров удалось выжить? Посчитайте долю выживших пассажиров. Ответ приведите в процентах, округлив до двух
знаков.

```python
sur_counts = data["Survived"].value_counts()
num_of_pas = sur_counts.get(0, 0) + sur_counts.get(1, 0)
frac_of_sur = sur_counts.get(1, 0) / num_of_pas * 100
print(round(frac_of_sur, 2))
```

3. Какую долю пассажиры первого класса составляли среди всех пассажиров? Ответ приведите в процентах, округлив до двух знаков.

```python
class_counts = data["Pclass"].value_counts()
frac_of_first = class_counts.get(1, 0) / num_of_pas * 100
print(round(frac_of_first, 2))
```

4. Какого возраста были пассажиры? Посчитайте среднее и медиану возраста пассажиров. В качестве ответа приведите два числа через пробел.

```python
age_mean = data["Age"].mean()
age_med = data["Age"].median()
print(round(age_mean, 2), age_med)
```

5. Коррелируют ли число братьев/сестер с числом родителей/детей?
Посчитайте корреляцию Пирсона между признаками SibSp и Parch.

```python
corr = data['SibSp'].corr(data['Parch'], method='pearson')
print(round(corr,2))
```

6. Какое самое популярное женское имя на корабле? Извлеките из
полного имени пассажира (колонка Name) его личное имя (First
Name). Это задание — типичный пример того, с чем сталкивает-
ся специалист по анализу данных. Данные очень разнородные и
шумные, но из них требуется извлечь необходимую информацию.
Попробуйте вручную разобрать несколько значений столбца Name
и выработать правило для извлечения имен, а также разделения
их на женские и мужские.

```python
def extract_first_name(name):
    if ',' not in name:
        return None

    after_comma = name.split(',', 1)[1].strip()
    female_titles = ['Mrs.', 'Miss.', 'Mlle.', 'Mme.']
    title = None

    for t in female_titles:
        if after_comma.startswith(t):
            title = t
            break
    if not title:
        return None

    rest = after_comma[len(title):].strip()

    if '(' in rest and ')' in rest:
        start = rest.find('(') + 1
        end = rest.find(')', start)
        if start < end:
            inside = rest[start:end]
            first_name = inside.split()[0].strip('"').strip("'").strip()
            return first_name

    words = rest.split()
    if words:
        first_name = words[0].strip('"').strip("'").strip()
        return first_name
    return None

women = data[data['Sex'] == 'female'].copy()
women['First Name'] = women['Name'].apply(extract_first_name)
valid_names = women['First Name'].dropna()
name_counts = valid_names.value_counts()
most_popular = name_counts.index[0] if not name_counts.empty else "None"

print(most_popular)
```
