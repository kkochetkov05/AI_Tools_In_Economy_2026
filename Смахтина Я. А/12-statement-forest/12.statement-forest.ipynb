```python
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import KFold, cross_val_score
from sklearn.metrics import r2_score
```

1. Загрузите данные из файла abalone.csv. Это датасет, в котором требуется предсказать возраст ракушки (число колец) по физическим измерениям.
2. Преобразуйте признак Sex в числовой: значение F должно перейти в -1, I — в 0, M — в 1.

```python
data = pd.read_csv('abalone.csv')

sex_mapping = {'F': -1, 'I': 0, 'M': 1}
data['Sex'] = data['Sex'].map(sex_mapping)
```

3. Разделите содержимое файлов на признаки и целевую переменную. В последнем столбце записана целевая переменная, в остальных — признаки.

```python
X = data.drop('Rings', axis=1).values
y = data['Rings'].values
```

4. Обучите случайный лес (sklearn.ensemble.RandomForestRegressor) с различным числом деревьев: от 1 до 50 (random_state=1). Для каждого из вариантов оцените качество работы полученного леса на кросс-валидации по 5 блокам. Используйте параметры "random_state=1" и "shuffle=True"при создании генератора кросс-валидации sklearn.cross_validation.KFold. В качестве меры качества воспользуйтесь коэффициентом детерминации (sklearn.metrics.r2_score).

```python
n_folds = 5
kf = KFold(n_splits=n_folds, shuffle=True, random_state=1)

n_trees_range = range(1, 51)
mean_r2_scores = []

for n in n_trees_range:
    rf = RandomForestRegressor(n_estimators=n, random_state=1, n_jobs=-1)
    scores = cross_val_score(rf, X, y, cv=kf, scoring='r2')
    mean_r2 = np.mean(scores)
    mean_r2_scores.append(mean_r2)

    if n % 10 == 0:
        print(f"n_estimators={n}, средний R² = {mean_r2:.4f}")
```

5. Определите, при каком минимальном количестве деревьев случайный лес показывает качество на кросс-валидации выше 0.52. Это количество и будет ответом на задание.

```python
min_trees = None
for n, r2 in zip(n_trees_range, mean_r2_scores):
    if r2 > 0.52:
        min_trees = n
        break

print(f"Минимальное количество деревьев: {min_trees}")
```

6. Обратите внимание на изменение качества по мере роста числа деревьев. Ухудшается ли оно?

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))
plt.plot(n_trees_range, mean_r2_scores, marker='o', linestyle='-', markersize=3)
plt.axhline(y=0.52, color='r', linestyle='--', label='R^2 = 0.52')
plt.axvline(x=min_trees, color='g', linestyle='--', label=f'min n = {min_trees}')
plt.xlabel('Количество деревьев')
plt.ylabel('Средний R^2 (5-fold CV)')
plt.title('Зависимость качества случайного леса от числа деревьев')
plt.legend()
plt.grid(True)
plt.show()

if all(mean_r2_scores[i] <= mean_r2_scores[i+1] + 0.005 for i in range(len(mean_r2_scores)-1)):
    print("Качество не ухудшается – оно растёт или стабилизируется.")
else:
    print("Наблюдаются колебания, но в целом тренд – улучшение или стабилизация.")
```
