```python
import numpy as np
import pandas as pd
from sklearn.metrics import roc_auc_score
```

1. Загрузите данные из файла data-logistic.csv. Это двумерная выборка, целевая переменная на которой принимает значения -1 или 1.

```python
data = pd.read_csv('data-logistic.csv', header=None)
X = data.iloc[:, 1:3].values
y = data.iloc[:, 0].values

print("Размер X:", X.shape)
print("Первые 5 меток:", y[:5])
```

2. Убедитесь, что выше выписаны правильные формулы для градиентного спуска.

3. Реализуйте градиентный спуск для обычной и L2-регуляризованной (с коэффициентом регуляризации 10) логистической регрессии. Используйте длину шага k=0.1. В качестве начального приближения используйте вектор (0, 0).

```python
def logistic_gradient_descent(X, y, C, k=0.1, tol=1e-5, max_iter=10000):

    n_samples, n_features = X.shape
    w = np.zeros(n_features)

    for i in range(max_iter):

        linear = np.dot(X, w)

        exponent = np.exp(-y * linear)
        sigma = 1 / (1 + exponent)

        grad = np.zeros(n_features)
        for j in range(n_features):
            grad[j] = (1.0 / n_samples) * np.sum(y * X[:, j] * (1 - sigma))

        grad_reg = grad - C * w
        w_new = w + k * grad_reg

        if np.linalg.norm(w_new - w) < tol:
            w = w_new
            break
        w = w_new
    return w
```

```python
w_no_reg = logistic_gradient_descent(X, y, C=0, k=0.1, tol=1e-5, max_iter=10000)
print("Веса без регуляризации:", w_no_reg)
```

4. Запустите градиентный спуск и доведите до сходимости (евклидово расстояние между векторами весов на соседних итерациях должно быть не больше 1e-5). Рекомендуется ограничить сверху число итераций десятью тысячами.

```python
w_reg = logistic_gradient_descent(X, y, C=10, k=0.1, tol=1e-5, max_iter=10000)
print("Веса с регуляризацией (C=10):", w_reg)
```

5. Какое значение принимает AUC-ROC на обучении без регуляризации и при ее использовании?

```python
def predict_proba(X, w):
    """Возвращает вероятность принадлежности к классу 1."""
    linear = np.dot(X, w)
    proba = 1 / (1 + np.exp(-linear))
    return proba

y_binary = (y + 1) // 2

proba_no_reg = predict_proba(X, w_no_reg)
auc_no_reg = roc_auc_score(y_binary, proba_no_reg)

proba_reg = predict_proba(X, w_reg)
auc_reg = roc_auc_score(y_binary, proba_reg)

print(f"AUC без регуляризации: {auc_no_reg:.6f}")
print(f"AUC с регуляризацией (C=10): {auc_reg:.6f}")
```
