```python
import pandas as pd
import numpy as np
from sklearn.decomposition import PCA
```

1. Загрузите данные close_prices.csv. В этом файле приведены цены акций 30 компаний на закрытии торгов за каждый день периода.

```python
df_stocks = pd.read_csv('close_prices.csv', parse_dates=['date'])
df_stocks.set_index('date', inplace=True)

df_dji = pd.read_csv('djia_index.csv', parse_dates=['date'])
df_dji.set_index('date', inplace=True)

print("Размер данных по акциям:", df_stocks.shape)
print("Размер данных по индексу:", df_dji.shape)
```

2. На загруженных данных обучите преобразование PCA с числом компоненты равным 10. Скольких компонент хватит, чтобы объяснить 90% дисперсии?

```python
pca = PCA(n_components=10)
pca.fit(df_stocks)

expl_var = pca.explained_variance_ratio_
print("Объяснённая дисперсия по компонентам (%):")
for i, val in enumerate(expl_var, 1):
    print(f"Компонента {i}: {val*100:.2f}%")
```

```python
cumsum = np.cumsum(expl_var)
n_components_90 = np.argmax(cumsum >= 0.90) + 1

print("Накопленная доля дисперсии по компонентам:")
for i, val in enumerate(cumsum, 1):
    print(f"  {i} компонент(а): {val*100:.2f}%")
print(f"\nДля объяснения 90% дисперсии достаточно {n_components_90} компонент(ы)")
```

3. Примените построенное преобразование к исходным данным и возьмите значения первой компоненты.

```python
X_pca = pca.transform(df_stocks)

first_component = X_pca[:, 0]

print("Первая главная компонента (первые 5 значений):")
print(first_component[:5])
```

4. Загрузите информацию об индексе Доу-Джонса из файла djia_prices.csv. Чему равна корреляция Пирсона между первой компонентой и индексом Доу-Джонса?

```python
dji_values = df_dji['^DJI'].values

correlation = np.corrcoef(first_component, dji_values)[0, 1]
print(f"Корреляция Пирсона: {correlation:.4f}")

print(f"Ответ: {correlation:.2f}")
```

5. Какая компания имеет наибольший вес в первой компоненте?


```python
loadings_first = pca.components_[0]
abs_loadings = np.abs(loadings_first)
max_idx = np.argmax(abs_loadings)
max_ticker = df_stocks.columns[max_idx]
max_weight = loadings_first[max_idx]

print(f"Компания с наибольшим весом в первой компоненте: {max_ticker}")
print(f"Вес (коэффициент): {max_weight:.4f}")
print(f"Абсолютное значение веса: {abs_loadings[max_idx]:.4f}")
```

```python
print("\n----- ответы -----")
print(f"1. {n_components_90}")
print(f"2. {correlation:.2f}")
print(f"3. {max_ticker}")
```
