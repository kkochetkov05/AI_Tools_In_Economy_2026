**Лабораторная работа №4**


```
import numpy as np
import sklearn
from sklearn.datasets import fetch_openml
from sklearn.preprocessing import scale
from sklearn.neighbors import KNeighborsRegressor
from sklearn.model_selection import cross_val_score, KFold
```

1. Загрузка набора данных Boston

```
boston = fetch_openml(name='boston', version=1, as_frame=True)
X = boston.data
y = boston.target

print("Размер матрицы признаков:", X.shape)
print("Целевой вектор:", y.shape)
```

2. Приводим признаки к одному масштабу

```
X_scaled = scale(X)

print("После масштабирования:")
print("Стандартное отклонение каждого признака:", np.std(X_scaled, axis=0))
```

3. Перебор параметра p метрики Минковского (от 1 до 10, 200 вариантов)

```
p_values = np.linspace(1, 10, 200)
mean_scores = []

# объект cv с перемешиванием
kf = KFold(n_splits=5, shuffle=True, random_state=42)

for p in p_values:
    knn = KNeighborsRegressor(n_neighbors=5,
                              weights='distance',
                              metric='minkowski',
                              p=p)

    scores = cross_val_score(knn, X_scaled, y,
                             cv=kf,
                             scoring='neg_mean_squared_error')

    mean_scores.append(scores.mean())
```

4. Поиск оптимального p


```
# ищем индекс максимального значения в mean_scores
best_idx = np.argmax(mean_scores)
best_p = p_values[best_idx]
best_score = mean_scores[best_idx]

print(f"Оптимальное значение p: {best_p:.4f}")
print(f"Лучшее среднее значение neg_mean_squared_error: {best_score:.4f}")
# обычная MSE = -best_score, тк `cross_val_score` возвращает отрицательную MSE
print(f"Соответствующая среднеквадратичная ошибка (MSE): {-best_score:.4f}")
```

Сохранение ответа в текстовый файл


```
answer = round(best_p, 1)
answer_str = str(answer)

with open('answer.txt', 'w') as f:
    f.write(answer_str)

print(f"Ответ сохранён в файл 'answer.txt'")
```
