```python
import pandas as pd
import numpy as np
from sklearn.linear_model import Perceptron
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import StandardScaler
```

```python
train = pd.read_csv('perceptron-train.csv', header=None)
test = pd.read_csv('perceptron-test.csv', header=None)
```

```python
X_train, y_train = train.iloc[:, 1:], train.iloc[:, 0]
X_test, y_test = test.iloc[:, 1:], test.iloc[:, 0]
```

```python
model = Perceptron(random_state=241)
model.fit(X_train, y_train)
pred = model.predict(X_test)
acc_before = accuracy_score(y_test, pred)
print(acc_before)
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
model.fit(X_train_scaled, y_train)
pred_scaled = model.predict(X_test_scaled)
acc_after = accuracy_score(y_test, pred_scaled)
print(acc_after)
```

```python
diff = acc_after - acc_before
print(round(diff, 3))

with open('ans.txt', 'w') as f:
    f.write(f"{diff:.3f}")
```
