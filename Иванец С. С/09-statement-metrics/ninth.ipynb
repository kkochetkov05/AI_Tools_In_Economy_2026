```python
import pandas as pd
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, precision_recall_curve
```

```python
df = pd.read_csv('classification.csv')
y_true = df['true']
y_pred = df['pred']
```

```python
true_positive = ((y_true == 1) & (y_pred == 1)).sum()
false_positive = ((y_true == 0) & (y_pred == 1)).sum()
false_negative = ((y_true == 1) & (y_pred == 0)).sum()
true_negative = ((y_true == 0) & (y_pred == 0)).sum()

print(f"ans1: {true_positive} {false_positive} {false_negative} {true_negative}")
with open('ans1.txt', 'w') as file:
    file.write(f"{true_positive} {false_positive} {false_negative} {true_negative}")
```

```python
accuracy_val = accuracy_score(y_true, y_pred)
precision_val = precision_score(y_true, y_pred)
recall_val = recall_score(y_true, y_pred)
f1_val = f1_score(y_true, y_pred)

with open('ans2.txt', 'w') as file:
    file.write(f"{accuracy_val:.2f} {precision_val:.2f} {recall_val:.2f} {f1_val:.2f}")

print(f"ans2: {accuracy_val:.2f} {precision_val:.2f} {recall_val:.2f} {f1_val:.2f}")
```

```python
data_frame = pd.read_csv('scores.csv')
ground_truth = data_frame['true']
model_scores = ['score_logreg', 'score_svm', 'score_knn', 'score_tree']
```

```python
roc_auc_values = {}
for model_name in model_scores:
    roc_auc_values[model_name] = roc_auc_score(ground_truth, data_frame[model_name])

top_model = max(roc_auc_values, key=lambda name: roc_auc_values[name])

with open('ans3.txt', 'w') as file:
    file.write(top_model)

print(f"ans3: {top_model} ({roc_auc_values[top_model]:.3f})")
```

```python
max_precision = 0
selected_classifier = None

for score_column in model_scores:
    prec_curve, rec_curve, _ = precision_recall_curve(ground_truth, data_frame[score_column])
    valid_indices = rec_curve >= 0.7
    
    if valid_indices.any():
        max_prec_for_model = prec_curve[valid_indices].max()
        if max_prec_for_model > max_precision:
            max_precision = max_prec_for_model
            selected_classifier = score_column

with open('ans4.txt', 'w') as file:
    file.write(f"{selected_classifier} {max_precision:.2f}")

print(f"ans4 : {selected_classifier} {max_precision:.2f}")
```
