```python
import pandas as pd
import re
```

### Загрузите датасет titanic.csv и, используя описанные выше способы работы с данными, найдите ответы на вопросы.


```python
data = pd.read_csv('titanic.csv', index_col='PassengerId')
```

### 1. Какое количество мужчин и женщин ехало на корабле? В качестве ответа приведите два числа через пробел.


```python
sex_counts = data['Sex'].value_counts()
print(sex_counts['male'], sex_counts['female'])
```

### 2. Какой части пассажиров удалось выжить? Посчитайте долю выживших пассажиров. Ответ приведите в процентах (число в интервале от 0 до 100, знак процента не нужен), округлив до двух знаков.


```python
survived_percent = 100 * data['Survived'].mean()
print(round(survived_percent, 2))
```

### 3. Какую долю пассажиры первого класса составляли среди всех пассажиров? Ответ приведите в процентах (число в интервале от 0 до 100, знак процента не нужен), округлив до двух знаков.


```python
first_class_percent = 100 * (data['Pclass'] == 1).mean()
print(round(first_class_percent, 2))
```

### 4. Какого возраста были пассажиры? Посчитайте среднее и медиану возраста пассажиров. В качестве ответа приведите два числа через пробел.


```python
mean_age = data['Age'].mean()
median_age = data['Age'].median()
print(round(mean_age, 2), median_age)
```

### 5. Коррелируют ли число братьев/сестер с числом родителей/детей? Посчитайте корреляцию Пирсона между признаками SibSp и Parch.


```python
corr = data['SibSp'].corr(data['Parch'])
print(round(corr, 2))
```

### 6. Какое самое популярное женское имя на корабле? Извлеките из полного имени пассажира (колонка Name) его личное имя (FirstName). Это задание — типичный пример того, с чем сталкивается специалист по анализу данных. Данные очень разнородные и шумные, но из них требуется извлечь необходимую информацию. Попробуйте вручную разобрать несколько значений столбца Name и выработать правило для извлечения имен, а также разделения их на женские и мужские.


```python
def extract_first_name(name):
    if 'Mrs.' in name:
        match = re.search(r'\((.*?)\)', name)
        if match:
            return match.group(1).split()[0]
        return name.split('Mrs.')[1].strip().split()[0]
    if 'Miss.' in name:
        return name.split('Miss.')[1].strip().split()[0]
    return None

female_names = data.loc[data['Sex'] == 'female', 'Name'].apply(extract_first_name)
print(female_names.value_counts().idxmax())
```
