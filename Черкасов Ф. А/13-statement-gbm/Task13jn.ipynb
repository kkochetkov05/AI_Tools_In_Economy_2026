```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.ensemble import GradientBoostingClassifier, RandomForestClassifier
from sklearn.metrics import log_loss
```

### 1. Загрузите выборку из файла gbm-data.csv с помощью pandas и преобразуйте ее в массив numpy (параметр values у датафрейма). 
##### В первой колонке файла с данными записано, была или нет реакция. Все остальные колонки (d1 - d1776) содержат различные характеристики молекулы, такие как размер, форма и т.д. Разбейте выборку на обучающую и тестовую, используя функцию train_test_split с параметрами test_size = 0.8 и random_state = 241.

```python
data = pd.read_csv('gbm-data.csv')
# Первая колонка - целевая переменная, остальные - признаки
y = data.iloc[:, 0].values
X = data.iloc[:, 1:].values

# Разбиваем выборку
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.8, random_state=241)
```

### 2. Обучите GradientBoostingClassifier с параметрами n_estimators=250, verbose=True, random_state=241 и для каждого значения learning_rate из списка [1, 0.5, 0.3, 0.2, 0.1] проделайте следующее:
##### • Используйте метод staged_decision_function для предсказания качества на обучающей и тестовой выборке на каждой итерации.
##### • Преобразуйте полученное предсказание по формуле $\frac{1}{1 + e^{-y_{pred}}}$, где $y_{pred}$ — предсказанное значение.
##### • Вычислите и постройте график значений log-loss на обучающей и тестовой выборках, а также найдите минимальное значение метрики и номер итерации, на которой оно достигается.

```python
# Список learning_rate для перебора
learning_rates = [1, 0.5, 0.3, 0.2, 0.1]

# Словари для хранения результатов
results = {}

for lr in learning_rates:
    print(f"\n--- Обучение с learning_rate = {lr} ---")
    
    # Создаем и обучаем модель
    gbc = GradientBoostingClassifier(n_estimators=250, learning_rate=lr, verbose=True, random_state=241)
    gbc.fit(X_train, y_train)
    
    # Получаем предсказания на каждой итерации
    # staged_decision_function возвращает сырые предсказания (не вероятности)
    train_pred_staged = gbc.staged_decision_function(X_train)
    test_pred_staged = gbc.staged_decision_function(X_test)
    
    # Преобразуем сырые предсказания в вероятности через формулу: p = 1 / (1 + exp(-y_pred))
    train_loss = []
    test_loss = []
    
    for train_pred, test_pred in zip(train_pred_staged, test_pred_staged):
        # Применяем формулу
        train_proba = 1.0 / (1.0 + np.exp(-train_pred.ravel()))
        test_proba = 1.0 / (1.0 + np.exp(-test_pred.ravel()))
        
        # Вычисляем log-loss
        train_loss.append(log_loss(y_train, train_proba))
        test_loss.append(log_loss(y_test, test_proba))
    
    # Находим минимальное значение метрики и номер итерации
    min_test_loss = min(test_loss)
    best_iter = test_loss.index(min_test_loss) + 1  # +1, т.к. итерации с 1
    
    # Сохраняем результаты
    results[lr] = {
        'train_loss': train_loss,
        'test_loss': test_loss,
        'min_test_loss': min_test_loss,
        'best_iter': best_iter
    }
    
    print(f"Минимальное значение log-loss на тесте: {min_test_loss:.2f}")
    print(f"Достигнуто на итерации: {best_iter}")
    
    # Строим график
    plt.figure(figsize=(10, 6))
    plt.plot(test_loss, 'r', linewidth=2, label='test')
    plt.plot(train_loss, 'g', linewidth=2, label='train')
    plt.legend()
    plt.title(f'Log-loss кривые (learning_rate = {lr})')
    plt.xlabel('Номер итерации')
    plt.ylabel('Log-loss')
    plt.grid(True, alpha=0.3)
    plt.show()
```

### 4. Приведите минимальное значение log-loss на тестовой выборке и номер итерации, на котором оно достигается, при learning_rate = 0.2.

```python
lr_02_result = results[0.2]

print(f"Минимальное значение log-loss на тесте: {lr_02_result['min_test_loss']:.2f}")
print(f"Номер итерации: {lr_02_result['best_iter']}")
```

### 5. На этих же данных обучите RandomForestClassifier с количеством деревьев, равным количеству итераций, на котором достигается наилучшее качество у градиентного бустинга из предыдущего пункта, random_state=241 и остальными параметрами по умолчанию.
##### Какое значение log-loss на тесте получается у этого случайного леса? (Не забывайте, что предсказания нужно получать с помощью функции predict_proba. В данном случае брать сигмоиду от оценки вероятности класса не нужно)


```python
# Берем количество деревьев, равное best_iter для learning_rate = 0.2
n_estimators_rf = lr_02_result['best_iter']

rf = RandomForestClassifier(n_estimators=n_estimators_rf, random_state=241)
rf.fit(X_train, y_train)

# Получаем предсказания вероятностей (predict_proba уже выдает вероятности)
# Для бинарной классификации берем вероятность второго класса (индекс 1)
rf_proba = rf.predict_proba(X_test)[:, 1]

# Вычисляем log-loss
rf_log_loss = log_loss(y_test, rf_proba)
print(f"Значение log-loss на тесте для случайного леса: {rf_log_loss:.2f}")
```
