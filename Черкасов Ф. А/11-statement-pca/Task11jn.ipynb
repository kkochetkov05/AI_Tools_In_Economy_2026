```python
import pandas as pd
import numpy as np
from sklearn.decomposition import PCA
```

### 1. Загрузите данные close_prices.csv. 
##### В этом файле приведены цены акций 30 компаний на закрытии торгов за каждый день периода.

```python
close_prices = pd.read_csv('close_prices.csv')
```

### 2. На загруженных данных обучите преобразование PCA с числом компоненты равным 10. Скольких компонент хватит, чтобы объяснить 90% дисперсии?


```python
# Убираем столбец с датами, оставляем только цены акций
X = close_prices.drop('date', axis=1)

# Обучение PCA с 10 компонентами
pca = PCA(n_components=10)
pca.fit(X)

# Определяем, сколько компонент нужно для объяснения 90% дисперсии
explained_variance_ratio = pca.explained_variance_ratio_
cumulative_variance = np.cumsum(explained_variance_ratio)

n_components_90 = np.where(cumulative_variance >= 0.9)[0][0] + 1
print(f"Для объяснения 90% дисперсии нужно {n_components_90} компонент")
```

### 3. Примените построенное преобразование к исходным данным и возьмите значения первой компоненты.

```python
X_transformed = pca.transform(X)
first_component = X_transformed[:, 0]
```

### 4. Загрузите информацию об индексе Доу-Джонса из файла djia_prices.csv. Чему равна корреляция Пирсона между первой компонентой и индексом Доу-Джонса?

```python
djia_index = pd.read_csv('djia_index.csv')

djia_values = djia_index['^DJI'].values

correlation = np.corrcoef(first_component, djia_values)[0, 1]
print(f"Корреляция Пирсона = {correlation:.2f}")
```

### 5. Какая компания имеет наибольший вес в первой компоненте?

```python
weights_first_component = pca.components_[0]

# Находим индекс компании с максимальным весом по модулю
max_weight_index = np.argmax(np.abs(weights_first_component))
max_weight_company = X.columns[max_weight_index]
max_weight_value = weights_first_component[max_weight_index]

print(f"Компания с наибольшим весом - {max_weight_company}")
```
