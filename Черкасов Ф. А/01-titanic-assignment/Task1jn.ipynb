```python
import pandas as pd
df = pd.read_csv('titanic.csv', index_col='PassengerId')
```

### 1. Какое количество мужчин и женщин ехало на корабле?
##### В качестве ответа приведите два числа через пробел.

```python
gender_counts = df['Sex'].value_counts()
male_count = gender_counts['male']
female_count = gender_counts['female']

print(male_count, female_count)
```

### 2. Какой части пассажиров удалось выжить? 
##### Посчитайте долю выживших пассажиров. Ответ приведите в процентах (число в интервале от 0 до 100, знак процента не нужен), округлив до двух знаков.

```python
survived_percent = (df['Survived'].sum() / len(df)) * 100
survived_percent_rounded = round(survived_percent, 2)

print(survived_percent_rounded)
```

### 3. Какую долю пассажиры первого класса составляли среди всех пассажиров? 
##### Ответ приведите в процентах (число в интервале от 0 до 100, знак процента не нужен), округлив до двух знаков.

```python
pclass_counts = df['Pclass'].value_counts()
first_class_percent = (pclass_counts[1] / len(df)) * 100
first_class_percent_rounded = round(first_class_percent, 2)

print(first_class_percent_rounded)
```

### 4. Какого возраста были пассажиры? 
##### Посчитайте среднее и медиа ну возраста пассажиров. Посчитайте среднее и медиану возраста пассажиров. В качестве ответа приведите два числа через пробел.

```python
mean_age = df['Age'].mean()
median_age = df['Age'].median()

# Округляем до двух знаков
mean_age_rounded = round(mean_age, 2)
median_age_rounded = round(median_age, 2)

print(mean_age_rounded, median_age_rounded)
```

### 5. Коррелируют ли число братьев/сестер с числом родителей/детей?
##### Посчитайте корреляцию Пирсона между признаками SibSp и Parch.

```python
correlation = df['SibSp'].corr(df['Parch'])
correlation_rounded = round(correlation, 2)

print(correlation_rounded)
```

### 6. Какое самое популярное женское имя на корабле? 
##### Извлеките из полного имени пассажира (колонка Name) его личное имя (First Name).

```python
# Сначала отфильтруем только женщин
women_df = df[df['Sex'] == 'female'].copy()

# Функция для извлечения имени из полного имени
def extract_woman_first_name(full_name):

    #Для Mrs. имя часто в скобках, для Miss./Mme./Lady. - после титула
    import re

    # Разделяем по запятой
    parts = full_name.split(',')
    if len(parts) < 2:
        return None
    
    name_part = parts[1].strip()
    
    # Проверяем, есть ли титул Mrs.
    if 'Mrs.' in name_part:
        # Ищем имя в скобках
        match = re.search(r'\(([^)]+)\)', name_part)
        if match:
            # Берем первое слово из скобок
            names_in_brackets = match.group(1).split()
            if names_in_brackets:
                # Убираем возможные кавычки
                return names_in_brackets[0].strip('"').strip("'")
    
    # Для Miss, Mme, Ms, Lady - берем имя после титула
    elif '.' in name_part:
        dot_pos = name_part.find('.')
        after_dot = name_part[dot_pos + 1:].strip()
        words = after_dot.split()
        if words:
            # Убираем скобки и кавычки
            first_name = words[0].strip('"').strip("'").strip('()')
            return first_name
    
    return None

# Применяем улучшенную функцию
women_df['FirstName'] = women_df['Name'].apply(extract_woman_first_name)

# Найдем самое популярное имя
name_counts = women_df['FirstName'].value_counts()
most_popular_name = name_counts.index[0]

print(most_popular_name)
```
