# Уменьшение количества цветов изображения

```python
import numpy as np
import matplotlib.pyplot as plt
from skimage.io import imread
from skimage import img_as_float
from sklearn.cluster import KMeans
```

1. Загрузите картинку parrots.jpg. Преобразуйте изображение, приведя все значения в интервал от 0 до 1.

```python
image = imread('parrots.jpg')
image = img_as_float(image)
w, h, d = image.shape
```

2. Создайте матрицу объекты-признаки.

```python
X = image.reshape(-1, 3)
```

```python
def psnr(img1, img2):
    mse = np.mean((img1 - img2) ** 2)
    if mse == 0:
        return 100
    return 10 * np.log10(1 / mse)
```

3. Запустите алгоритм K-Means.

```python
def compress_image(labels, centers, method):
    new_img = np.zeros_like(X)
    for i in range(len(centers)):
        cluster_pixels = X[labels == i]
        if method == 'mean':
            color = cluster_pixels.mean(axis=0)
        else:
            color = np.median(cluster_pixels, axis=0)
        new_img[labels == i] = color
    return new_img.reshape(w, h, d)
```

```python
best_k = None
for k in range(1, 21):
    kmeans = KMeans(n_clusters=k, init='k-means++', random_state=241)
    labels = kmeans.fit_predict(X)
    centers = kmeans.cluster_centers_
    img_mean = compress_image(labels, centers, 'mean')
    psnr_mean = psnr(image, img_mean)
    img_median = compress_image(labels, centers, 'median')
    psnr_median = psnr(image, img_median)
    if psnr_mean > 20 or psnr_median > 20:
        best_k = k
        break
print(best_k)
```
