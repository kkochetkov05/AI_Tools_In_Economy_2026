# Метрики качества классификации

```python
import pandas as pd
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.metrics import roc_auc_score, precision_recall_curve
```

```python
data = pd.read_csv('classification.csv')
y_true = data['true']
y_pred = data['pred']
```

```python
TP = ((y_true == 1) & (y_pred == 1)).sum()
FP = ((y_true == 0) & (y_pred == 1)).sum()
FN = ((y_true == 1) & (y_pred == 0)).sum()
TN = ((y_true == 0) & (y_pred == 0)).sum()
print("TP FP FN TN:")
print(TP, FP, FN, TN)
```

```python
accuracy = accuracy_score(y_true, y_pred)
precision = precision_score(y_true, y_pred)
recall = recall_score(y_true, y_pred)
f1 = f1_score(y_true, y_pred)
print("Metrics:")
print("Accuracy:", accuracy)
print("Precision:", precision)
print("Recall:", recall)
print("F1:", f1)
```

```python
scores = pd.read_csv('scores.csv')
y_true_scores = scores['true']
```

```python
roc_logreg = roc_auc_score(y_true_scores, scores['score_logreg'])
roc_svm = roc_auc_score(y_true_scores, scores['score_svm'])
roc_knn = roc_auc_score(y_true_scores, scores['score_knn'])
roc_tree = roc_auc_score(y_true_scores, scores['score_tree'])
roc_scores = {'score_logreg': roc_logreg, 'score_svm': roc_svm, 'score_knn': roc_knn, 'score_tree': roc_tree}
best_roc = max(roc_scores, key=roc_scores.get)
print("Best ROC-AUC:", best_roc, roc_logreg)
```

```python
best_precision = {}
for col in ['score_logreg', 'score_svm', 'score_knn', 'score_tree']:
    precision_arr, recall_arr, thresholds = precision_recall_curve(y_true_scores, scores[col])
    valid = precision_arr[recall_arr >= 0.7]
    best_precision[col] = max(valid)
best_model = max(best_precision, key=best_precision.get)
print(best_model, round(best_precision[best_model], 2))
```
