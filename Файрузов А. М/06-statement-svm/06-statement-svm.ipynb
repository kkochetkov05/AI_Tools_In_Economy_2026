# SVM

```python
import pandas as pd
import numpy as np
from sklearn.svm import SVC
```

```python
data = pd.read_csv('svm-data.csv', header=None)
```

```python
y = data.iloc[:, 0]
X = data.iloc[:, 1:]
```

```python
clf = SVC(kernel='linear', C=100000, random_state=241)
clf.fit(X, y)
support_indices = clf.support_
result = support_indices + 1
print(', '.join(map(str, sorted(result))))
```
