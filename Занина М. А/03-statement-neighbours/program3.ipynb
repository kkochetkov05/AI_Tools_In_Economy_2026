# 1. Загрузите выборку Wine по адресу https://archive.ics.uci.edu/ml/machinelearning-databases/wine/wine.data

```python
import pandas as pd
import numpy as np

data_url = "https://archive.ics.uci.edu/ml/machine-learning-databases/wine/wine.data"
raw_df = pd.read_csv(data_url, header=None)
raw_df.head()
```

# 2. Извлеките из данных признаки и классы. Класс записан в первом столбце (три варианта), признаки в столбцах со второго по последний. Более подробно о сути признаков можно прочитать по адресу https://archive.ics.uci.edu/ml/datasets/Wine (см. также файл wine.names, приложенный к заданию)

```python
data = raw_df.values
X = raw_df.iloc[:, 1:]
y = raw_df.iloc[:, 0]
```

# 3. Оценку качества необходимо провести методом кросс-валидации по 5 блокам (5-fold). Создайте генератор разбиений, который переме шивает выборку перед формированием блоков (shuffle=True). Для воспроизводимости результата, создавайте генератор KFold с фиксированным параметром random_state=42. В качестве меры качества используйте долю верных ответов (accuracy)

```python
from sklearn.model_selection import KFold

kf = KFold(n_splits=5, shuffle=True, random_state=42)
```

# 4. Найдите точность классификации на кросс-валидации для метода k ближайших соседей (sklearn.neighbors.KNeighborsClassifier), при k от 1 до 50. При каком k получилось оптимальное качество? Чему оно равно (число в интервале от 0 до 1)? Данные результаты и будут ответами на вопросы 1 и 2.

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import cross_val_score

k_opt = 1
score_max = 0
for k in range (1, 51):
    model = KNeighborsClassifier(n_neighbors = k)
    scores = cross_val_score(model, X, y, cv = kf, scoring = 'accuracy')
    if scores.mean() > score_max:
        score_max = scores.mean()
        k_opt = k
print("Лучшее качество:", round (score_max, 2))
print("Оптимальное k:", k_opt)
```

# 5. Произведите масштабирование признаков с помощью функции sklearn.preprocessing.scale. Снова найдите оптимальное k на кросс валидации.

```python
from sklearn.preprocessing import scale

X_scaled = scale(X)

k_opt_scale = 1
score_max_scale = 0

for k in range (1, 51):
    model = KNeighborsClassifier(n_neighbors = k)
    scores = cross_val_score(model, X_scaled, y, cv = kf, scoring = 'accuracy')

    if scores.mean() > score_max_scale:
        score_max_scale = scores.mean()
        k_opt_scale = k

print("Оптимальное k:", k_opt_scale)
#print("Лучшее качество:", round(score_max_scale, 2))
```

# 6. Какое значение k получилось оптимальным после приведения признаков к одному масштабу? Как изменилось значение качества? Приведите ответы на вопросы 3 и 4.


```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import cross_val_score
from sklearn.preprocessing import scale

k_opt = 1
score_max = 0

for k in range(1, 51):
    model = KNeighborsClassifier(n_neighbors=k)
    scores = cross_val_score(model, X, y, cv=kf, scoring='accuracy')
    if scores.mean() > score_max:
        score_max = scores.mean()
        k_opt = k

X_scaled = scale(X)

k_opt_scaled = 1
score_max_scaled = 0

for k in range(1, 51):
    model = KNeighborsClassifier(n_neighbors=k)
    scores = cross_val_score(model, X_scaled, y, cv=kf, scoring='accuracy')
    if scores.mean() > score_max_scaled:
        score_max_scaled = scores.mean()
        k_opt_scaled = k

print(f"До масштабирования:    k = {k_opt}, качество = {round(score_max, 2)}")
print(f"После масштабирования: k = {k_opt_scaled}, качество = {round(score_max_scaled, 2)}")
```
