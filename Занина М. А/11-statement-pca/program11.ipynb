# 1. Загрузите данные close_prices.csv. В этом файле приведены цены акций 30 компаний на закрытии торгов за каждый день периода.

```python
import pandas as pd
import numpy as np
from sklearn.decomposition import PCA

data = pd.read_csv("/content/close_prices.csv")
data_for_pca = data.iloc[:, 1:]
```

# 2. На загруженных данных обучите преобразование PCA с числом компоненты равным 10. Скольких компонент хватит, чтобы объяснить 90% дисперсии?

```python
pca = PCA(n_components=10)
data_pca = pca.fit_transform(data_for_pca)

explained = pca.explained_variance_ratio_
cum_explained = np.cumsum(explained)
n_components_90 = np.argmax(cum_explained >= 0.9) + 1

print(int(n_components_90))
```

# 3. Примените построенное преобразование к исходным данным и возьмите значения первой компоненты.

```python
pc1 = data_pca[:, 0]
```

# 4. Загрузите информацию об индексе Доу-Джонса из файла djia_prices.csv. Чему равна корреляция Пирсона между первой компонентой и индексом Доу-Джонса?

```python
d_j_index = pd.read_csv("/content/djia_index.csv")
dji = d_j_index['^DJI'].values

corr_matrix = np.corrcoef(pc1, dji)
pearson_corr = corr_matrix[0, 1]

print(round(pearson_corr, 2))
```

# 5. Какая компания имеет наибольший вес в первой компоненте?

```python
weights = pca.components_[0]

j = np.argmax(weights)
company = data_for_pca.columns[j]

print(company)
```
