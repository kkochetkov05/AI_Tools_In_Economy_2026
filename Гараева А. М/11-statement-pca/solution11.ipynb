## Корреляция Пирсона: 0.91
## В первой компоненте наибольший вес имеет компания: V (0.58)

```python
import pandas as pd
import numpy as np
from sklearn.decomposition import PCA
```

```python
data = pd.read_csv('close_prices.csv')
d_j_index = pd.read_csv('djia_index.csv')
```

```python
data.head()
```

```python
d_j_index.head()
```

```python
data_for_pca = data.iloc[:, 1:]
data_for_pca.head()
```

```python
pca = PCA(n_components=10)
data_pca = pca.fit_transform(data_for_pca)
```

```python
explained = pca.explained_variance_ratio_
cum_explained = np.cumsum(explained)
n_components_90 = np.argmax(cum_explained >= 0.9) + 1
int(n_components_90)
```

```python
pc1 = data_pca[:, 0]
dji = d_j_index['^DJI'].values

corr_matrix = np.corrcoef(pc1, dji)
pearson_corr = corr_matrix[0, 1]

print(round(pearson_corr, 2))
```

```python
w = pca.components_[0]
j = np.argmax(np.abs(w))
company = data_for_pca.columns[j]
print(company, round(w[j], 2))
```
