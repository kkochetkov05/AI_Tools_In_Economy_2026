6. Какое самое популярное женское имя на корабле? Извлеките из полного имени пассажира (колонка Name) его личное имя (First Name).

```python
female = data[data['Sex'] == 'female'].copy()

import re

def extract_first_name(name):

    m = re.search(r'\((.*?)\)', name)
    if m:
        return m.group(1).split()[0]

    m = re.search(r'\.\s*([A-Za-z]+)', name)
    if m:
        return m.group(1)

    return None

female['FirstName'] = female['Name'].apply(extract_first_name)
```
