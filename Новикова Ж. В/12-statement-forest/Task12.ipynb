```python
import pandas as pd
```

```python
data = pd.read_csv("abalone.csv")
data.head(10)
```

```python
data['Sex'] = data['Sex'].map(lambda x: 1 if x == 'M' else (-1 if x == 'F' else 0))
X = data.iloc[:, :-1]
Y = data.iloc[:, -1]
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import KFold, cross_val_score

kf = KFold(n_splits=5, shuffle=True, random_state=1)

scores = []

for n_trees in range(1, 51):
    rf = RandomForestRegressor(n_estimators=n_trees, random_state=1)
    score = cross_val_score(rf, X, Y, cv=kf, scoring='r2').mean()
    scores.append(score)

```

Определите, при каком минимальном количестве деревьев случайный лес показывает качество на кросс-валидации выше 0.52.
Это количество и будет ответом на задание.


```python
for i, score in enumerate(scores):
    if score > 0.52:
        print(i + 1)
        break
```

Обратите внимание на изменение качества по мере роста числа деревьев. Ухудшается ли оно?
Отыет: нет, не ухудшается, сначала можно заметить быстрый рост, а потом значения качества стабилизируются
