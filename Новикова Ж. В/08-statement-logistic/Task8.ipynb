```python
import pandas as pd
import numpy as np
from sklearn.metrics import roc_auc_score
```

```python
df = pd.read_csv("data_logistic.csv", header = None)
y = df[0].values
X = df[[1, 2]].values
```

Какое значение принимает AUC-ROC на обучении без регуляризации?

```python
l = len(y)
k = 0.1
C = 0      # без регуляризации
w1, w2 =2, 5

for i in range(10000):
    w1_old, w2_old = w1, w2

    sum1 = 0
    sum2 = 0

    for i in range(l):
        a = y[i] * (w1 * X[i,0] + w2 * X[i,1])
        grad = y[i] * (1 - 1/(1 + np.exp(-a)))

        sum1 += X[i,0] * grad
        sum2 += X[i,1] * grad

    w1 += k * (sum1 / l - C * w1)
    w2 += k * (sum2 / l - C * w2)

    if np.sqrt((w1 - w1_old)**2 + (w2 - w2_old)**2) < 1e-5:
        break

# вероятности
probs = 1 / (1 + np.exp(-(w1 * X[:,0] + w2 * X[:,1])))
auc_no_reg = roc_auc_score(y, probs)
print(round(auc_no_reg, 3))
```

Какое значение принимает AUC-ROC на обучении c регуляризацией?

```python
C = 10
w1, w2 = 0, 0

for i in range(10000):
    w1_old, w2_old = w1, w2

    sum1 = 0
    sum2 = 0

    for i in range(l):
        a = y[i] * (w1 * X[i,0] + w2 * X[i,1])
        grad = y[i] * (1 - 1/(1 + np.exp(-a)))

        sum1 += X[i,0] * grad
        sum2 += X[i,1] * grad

    w1 += k * (sum1 / l - C * w1)
    w2 += k * (sum2 / l - C * w2)

    if np.sqrt((w1 - w1_old)**2 + (w2 - w2_old)**2) < 1e-5:
        break

# вероятности
probs = 1 / (1 + np.exp(-(w1 * X[:,0] + w2 * X[:,1])))
auc_reg = roc_auc_score(y, probs)
print(round(auc_reg, 3))
```
