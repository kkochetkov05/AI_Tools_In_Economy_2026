```python
import pandas as pd
import numpy as np
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, precision_recall_curve
```

```python
df = pd.read_csv('classification.csv')
print(df.head())
print(df.info())
print(df.describe())
```

Подсчитайте величины TP, FP, FN и TN согласно их определениям. Ответ в данном вопросе — четыре числа через пробел.

```python
TP = len(df[(df.true == 1) & (df.pred == 1)])
FP = len(df[(df.true == 0) & (df.pred == 1)])
FN = len(df[(df.true == 1) & (df.pred == 0)])
TN = len(df[(df.true == 0) & (df.pred == 0)])

print(TP, FP, FN, TN)
```

Посчитайте основные метрики качества классификатора:
• Accuracy
• Precision
• Recall
• F-мера

```python
y_true = df.true
y_pred = df.pred

print(round(accuracy_score(y_true, y_pred), 2))
print(round(precision_score(y_true, y_pred), 2))
print(round(recall_score(y_true, y_pred), 2))
print(round(f1_score(y_true, y_pred), 2))
```

Посчитайте площадь под ROC-кривой для каждого классификатора. Какой классификатор имеет наибольшее значение метрики
AUC-ROC (укажите название столбца с ответами этого классификатора)?

```python
scores = pd.read_csv("scores.csv")

y_true = scores.true

auc_logreg = roc_auc_score(y_true, scores.score_logreg)
auc_svm = roc_auc_score(y_true, scores.score_svm)
auc_knn = roc_auc_score(y_true, scores.score_knn)
auc_tree = roc_auc_score(y_true, scores.score_tree)

value = {'auc_logreg' : round(auc_logreg, 2), 'auc_svm' : round(auc_svm, 2), "auc_knn" : round(auc_knn, 2), "auc_tree" : round(auc_tree, 2) }

print(max(value, key=value.get))
print(value[max(value, key=value.get)])
# print(max(auc_logreg, auc_svm, auc_knn, auc_tree))
```

Какой классификатор достигает наибольшей точности (Precision) при полноте (Recall) не менее 70% (укажите название столбца с
ответами этого классификатора)?
Какое значение точности при этом получается?

```python
for col in ['score_logreg', 'score_svm', 'score_knn', 'score_tree']:

    precision, recall, thresholds = precision_recall_curve(y_true, scores[col])
    good_precision = precision[recall >= 0.7]
    best_precision = round(max(good_precision), 2)

print(col, best_precision)
```
