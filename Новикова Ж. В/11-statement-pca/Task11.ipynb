```python
import pandas as pd
import numpy as np
from sklearn.decomposition import PCA
```

```python
price = pd.read_csv("close_prices.csv")
index = pd.read_csv("djia_index.csv")
print(price.head())
print(index.head())
```

На загруженных данных обучите преобразование PCA с числом компоненты равным 10.
Скольких компонент хватит, чтобы объяснить 90% дисперсии? Ответ: 4х компонент хватит

```python
X = price.iloc[:, 1:]
pca = PCA(n_components=10)
pca.fit(X)

print(pca.explained_variance_ratio_)

# сколько дисперсии объясняют первые N компонент вместе
print(np.cumsum(pca.explained_variance_ratio_))
```

Примените построенное преобразование к исходным данным и возьмите значения первой компоненты.
Загрузите информацию об индексе Доу-Джонса из файла djia_prices.csv.
Чему равна корреляция Пирсона между первой компонентой и индексом Доу-Джонса?

```python
X_pca = pca.transform(X)
first_component = X_pca[:,0]

index_values = index.iloc[:, 1]

corr = np.corrcoef(first_component, index_values)[0, 1]
print(round(corr, 2))
```

Какая компания имеет наибольший вес в первой компоненте?

```python
weights = pca.components_[0]
max_weight_index = np.argmax(np.abs(weights))
print(X.columns[max_weight_index])
```
