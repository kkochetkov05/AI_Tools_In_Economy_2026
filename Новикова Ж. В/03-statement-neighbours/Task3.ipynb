```python
import pandas as pd
import numpy as np
from sklearn.model_selection import KFold, cross_val_score
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import scale
```

```python
df = pd.read_csv('winedata')
```

Извлеките из данных признаки и классы. Класс записан в первом столбце (три варианта), признаки — в столбцах со второго по последний.

```python
y = df.iloc[:, 0]
X = df.iloc[:, 1:]
```

Найдите точность классификации на кросс-валидации для метода k ближайших соседей (sklearn.neighbors.KNeighborsClassifier), при k от 1 до 50. При каком k получилось оптимальное качество? Чему оно равно (число в интервале от 0 до 1)? Данные результаты и будут ответами на вопросы 1 и 2.

```python
kf = KFold(n_splits=5, shuffle=True, random_state=42)

scores = []
for k in range(1, 50):
    model = KNeighborsClassifier(n_neighbors=k)
    cv_scores = cross_val_score(model, X, y, cv=kf, scoring='accuracy')
    scores.append(cv_scores.mean())

best_k = np.argmax(scores) + 1
best_score = max(scores)

print(best_k)
print(round(best_score, 2))
```

Произведите масштабирование признаков с помощью функции sklearn.preprocessing.scale. Снова найдите оптимальное k на кроссвалидации.
Какое значение k получилось оптимальным после приведения признаков к одному масштабу? Как изменилось значение качества?

```python
X_scaled = scale(X)

scores_scaled = []
for k in range(1, 50):
    model = KNeighborsClassifier(n_neighbors=k)
    cv_scores = cross_val_score(model, X_scaled, y, cv=kf, scoring='accuracy')
    scores_scaled.append(cv_scores.mean())

best_k_scaled = np.argmax(scores_scaled) + 1
best_score_scaled = max(scores_scaled)

print(best_k_scaled)
print(round(best_score_scaled, 2))
```
