# 1. Какие признаки нужно взять для построения дерева?

Используем признаки `Pclass`, `Fare`, `Age` и `Sex`, а целевой переменной будет `Survived`.


```python
import pandas as pd
from sklearn import tree
data =pd.read_csv('titanic.csv')
X=data[['Pclass', 'Fare', 'Age', 'Sex']].copy()
y=data['Survived'].copy()
X['Sex'] = X['Sex'].map({'male': 0, 'female': 1})

mask=X.notna().all(axis=1)
X_clean = X[mask]
y_clean = y[mask]
```

# 2. Постройте решающее дерево и найдите важности признаков

Постройте `DecisionTreeClassifier(random_state=241)` и посмотрите на `feature_importances_`.


```python
clf = tree.DecisionTreeClassifier(random_state=241)
clf = clf.fit(X_clean, y_clean)
tree.plot_tree(clf)

imp=clf.feature_importances_
X_names = X_clean.columns

importance_pairs = list(zip(X_names, imp))
importance_pairs.sort(key=lambda x: x[1], reverse=True)
```

# 3. Какие два признака оказались наиболее важными?

В ответе нужно указать два признака с наибольшей важностью.


```python
top_two = [importance_pairs[0][0], importance_pairs[1][0]]
print(','.join(top_two))
```
