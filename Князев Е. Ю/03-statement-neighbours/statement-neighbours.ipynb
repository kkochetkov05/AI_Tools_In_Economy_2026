## 1. Загрузите выборку Wine по адресу https://archive.ics.uci.edu/ml/machine-learning-databases/wine/wine.data

```python
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import KFold, cross_val_score
from sklearn.preprocessing import scale

data = pd.read_csv('wine.data', header=None)
```

## 2. Извлеките из данных признаки и классы.

Класс записан в первом столбце (три варианта), признаки — в столбцах со второго по последний.

```python
classes = data[0]
features = data.iloc[:, 1:]
```

## 3. Оценку качества необходимо провести методом кросс-валидации по 5 блокам (5-fold).

Создайте генератор разбиений, который перемешивает выборку перед формированием блоков (`shuffle=True`). Для воспроизводимости результата создавайте генератор `KFold` с фиксированным параметром `random_state=42`. В качестве меры качества используйте долю верных ответов (accuracy).

```python
kf = KFold(n_splits=5, shuffle=True, random_state=42)
k_scores = []
results = pd.DataFrame(columns=['k', 'accuracy'])
```

## 4. Найдите точность классификации на кросс-валидации для метода k ближайших соседей при k от 1 до 50.

При каком k получилось оптимальное качество? Чему оно равно (число в интервале от 0 до 1)? Данные результаты и будут ответами на вопросы 1 и 2.

```python
for k in range(1, 51):
    model = KNeighborsClassifier(n_neighbors=k)
    k_scores.append(cross_val_score(model, features, classes, cv=kf, scoring='accuracy').mean())

best_k = range(1, 51)[k_scores.index(max(k_scores))]
best_accuracy = max(k_scores)
```

## 5. Произведите масштабирование признаков с помощью функции `sklearn.preprocessing.scale`. Снова найдите оптимальное k на кросс-валидации.

```python
features_scaled = scale(features)

k_scores_scaled = []

for k in range(1, 51):
    model = KNeighborsClassifier(n_neighbors=k)
    score = cross_val_score(model, features_scaled, classes, cv=kf, scoring='accuracy').mean()
    k_scores_scaled.append(score)

best_k_scaled = range(1, 51)[k_scores_scaled.index(max(k_scores_scaled))]
best_accuracy_scaled = max(k_scores_scaled)
```

## 6. Какое значение k получилось оптимальным после приведения признаков к одному масштабу? Как изменилось значение качества?

Приведите ответы на вопросы 3 и 4.

```python
print("\nСРАВНЕНИЕ:")
print(f"Без масштабирования: k={best_k}, accuracy={best_accuracy:.4f}")
print(f"С масштабированием:  k={best_k_scaled}, accuracy={best_accuracy_scaled:.4f}")
```
