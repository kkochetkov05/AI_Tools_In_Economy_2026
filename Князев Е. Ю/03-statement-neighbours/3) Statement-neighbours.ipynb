# 1. Загрузите выборку `wine.data`

В первом столбце находится класс вина, в остальных 13 столбцах признаки объекта.


```python
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import KFold, cross_val_score
from sklearn.preprocessing import scale

data=pd.read_csv('wine.data', header=None)
classes=data[0]
features=data.iloc[:,1:]

kf= KFold(n_splits=5, shuffle=True,random_state=42)
k_scores=[]
results = pd.DataFrame(columns=['k', 'accuracy'])
```

# 2. Найдите лучший `k` для kNN без масштабирования

Переберите `k` от 1 до 50 и вычислите среднее качество по кросс-валидации.


```python
for k in range(1,51):
    model=KNeighborsClassifier(n_neighbors=k)
    k_scores.append(cross_val_score(model, features, classes, cv=kf, scoring='accuracy').mean())

best_k = range(1, 51)[k_scores.index(max(k_scores))]
best_accuracy = max(k_scores)
```

# 3. Масштабируйте признаки и снова найдите лучший `k`

Используйте масштабирование и повторите подбор параметра `k`.


```python
features_scaled=scale(features)

k_scores_scaled = []

for k in range(1, 51):
    model = KNeighborsClassifier(n_neighbors=k)
    score = cross_val_score(model, features_scaled, classes, cv=kf, scoring='accuracy').mean()
    k_scores_scaled.append(score)

best_k_scaled = range(1, 51)[k_scores_scaled.index(max(k_scores_scaled))]
best_accuracy_scaled = max(k_scores_scaled)
```

# 4. Ответ

Сравните качество до и после масштабирования признаков.


```python
print("\nСРАВНЕНИЕ:")
print(f"Без масштабирования: k={best_k}, accuracy={best_accuracy:.4f}")
print(f"С масштабированием:  k={best_k_scaled}, accuracy={best_accuracy_scaled:.4f}")
```
