# 1. Посчитайте элементы матрицы ошибок

Найдите значения `TP`, `FP`, `FN`, `TN`.


```python
import pandas as pd
from sklearn.metrics import accuracy_score, precision_score,recall_score, f1_score,roc_auc_score, precision_recall_curve

clf = pd.read_csv('classification.csv')
y_true = clf['true']
y_pred = clf['pred']
TP = ((y_true==1) & (y_pred==1)).sum()
FP = ((y_true==0) & (y_pred==1)).sum()
FN = ((y_true==1) & (y_pred==0)).sum()
TN = ((y_true==0) & (y_pred==0)).sum()
print(TP, FP, TN, FN)
```

# 2. Посчитайте основные метрики качества

Вычислите `accuracy`, `precision`, `recall` и `f1`.


```python
accuracy = accuracy_score(y_true, y_pred)
precision = precision_score(y_true, y_pred)
recall = recall_score(y_true, y_pred)
f1 = f1_score(y_true, y_pred)
print(f"{accuracy:.2f} {precision:.2f} {recall:.2f} {f1:.2f}")
```

# 3. Какой классификатор имеет наибольшее значение ROC-AUC?

Проверьте столбцы `score_logreg`, `score_svm`, `score_knn`, `score_tree`.


```python
scores = pd.read_csv('scores.csv')
y_true2 = scores['true']
score_cols = ['score_logreg', 'score_svm', 'score_knn', 'score_tree']

for col in score_cols:
    auc = roc_auc_score(y_true2,scores[col] )
    print(col, auc)
```

# 4. Какой классификатор дает наибольшую precision при recall >= 0.7?

Для каждой модели найдите максимальную precision при условии `recall >= 0.7`.


```python
for col in score_cols:
    precision_arr, recall_arr, _ = precision_recall_curve(y_true2,scores[col] )
    mask = recall_arr >= 0.7
    best = precision_arr[mask].max()
    print(col, best)
```
