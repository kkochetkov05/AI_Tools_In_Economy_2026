## 1. Какое количество мужчин и женщин ехало на корабле?

Загрузите датасет `titanic.csv`. В качестве ответа приведите два числа через пробел.

```python
import pandas as pd
data =pd.read_csv('titanic.csv', index_col='PassengerId')
passenger_count=len(data)
#1
n=data['Sex'].value_counts()
male, female = n['male'],n['female']
print('1)',male, female)
```

## 2. Какой части пассажиров удалось выжить?

Посчитайте долю выживших пассажиров. Ответ приведите в процентах (число в интервале от 0 до 100, знак процента не нужен), округлив до двух знаков.

```python
#2
Survived=data['Survived'].value_counts()
print('2)',f"{Survived[1]/passenger_count*100:.2f}")
```

## 3. Какую долю пассажиры первого класса составляли среди всех пассажиров?

Ответ приведите в процентах (число в интервале от 0 до 100, знак процента не нужен), округлив до двух знаков.

```python
#3
Pclass=data['Pclass'].value_counts()
print('3)',f"{Pclass[1]/passenger_count*100:.2f}")
```

## 4. Какого возраста были пассажиры?

Посчитайте среднее и медиану возраста пассажиров. В качестве ответа приведите два числа через пробел.

```python
#4
mean_age = data['Age'].mean()
median_age = data['Age'].median()
answer4 = f"{mean_age:.2f} {median_age:.2f}"
print('4)', answer4)
```

## 5. Коррелируют ли число братьев/сестер с числом родителей/детей?

Посчитайте корреляцию Пирсона между признаками `SibSp` и `Parch`.

```python
#5
Cor=data['SibSp'].corr(data['Parch'])
print('5)',f"{Cor:.2f}")
```

## 6. Какое самое популярное женское имя на корабле?

Извлеките из полного имени пассажира (колонка `Name`) его личное имя (First Name). Попробуйте вручную разобрать несколько значений столбца `Name` и выработать правило для извлечения имен, а также разделения их на женские и мужские.

```python
#6
def female_name(name):
    female_titles=['Mrs.', 'Miss.', 'Ms.', 'Mme.', 'Lady.']
    for title in female_titles:
        if title in name:
            return True
    return False

def female_first_name(name):
    female_titles = ['Miss.']
    for title in female_titles:
        if title in name:
            after_title = name.split(title)[1].strip()
            first_name = after_title.split()[0]
            first_name = first_name.strip('.')
            return first_name
    return None
female_names=[]
for name in data['Name']:
    result= female_first_name(name)
    if result:
        female_names.append(result)

name_counts={}
for name in female_names:
    name_counts[name]=name_counts.get(name, 0) + 1
popular=max(name_counts, key=name_counts.get)
print(popular)
```
