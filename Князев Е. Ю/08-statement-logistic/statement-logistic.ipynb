## 3. Реализуйте градиентный спуск для обычной и L2-регуляризованной (с коэффициентом регуляризации 10) логистической регрессии.

Используйте длину шага `k=0.1`. В качестве начального приближения используйте вектор (0, 0).

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def gradient_descent(X, y, C=0, k=0.1, max_iter=10000):
    w1, w2 = 0.0, 0.0
    x1 = X[1].values
    x2 = X[2].values
    for _ in range(max_iter):
        pred = y * (w1 * x1 + w2 * x2)
        error = 1 - sigmoid(pred)
        w1_new = w1 + k * np.mean(y * x1 * error) - k * C * w1
        w2_new = w2 + k * np.mean(y * x2 * error) - k * C * w2
        if np.sqrt((w1_new - w1) ** 2 + (w2_new - w2) ** 2) < 1e-5:
            break
        w1, w2 = w1_new, w2_new
    return w1, w2
```

## 4. Запустите градиентный спуск и доведите до сходимости.

Евклидово расстояние между векторами весов на соседних итерациях должно быть не больше 1e-5. Рекомендуется ограничить сверху число итераций десятью тысячами.

```python
w1, w2 = gradient_descent(X, y, C=0)
w1_reg, w2_reg = gradient_descent(X, y, C=10)
x1 = X[1].values
x2 = X[2].values
```

## 5. Какое значение принимает AUC-ROC на обучении без регуляризации и при ее использовании?

Эти величины будут ответом на задание. В качестве ответа приведите два числа через пробел. Обратите внимание, что на вход функции `roc_auc_score` нужно подавать оценки вероятностей, подсчитанные обученным алгоритмом. Для этого воспользуйтесь сигмоидной функцией: `a(x) = 1/(1 + exp(−w1x1 − w2x2))`.

```python
auc = roc_auc_score(y, sigmoid(w1 * x1 + w2 * x2))
auc_reg = roc_auc_score(y, sigmoid(w1_reg * x1 + w2_reg * x2))

print(f"{auc:.3f} {auc_reg:.3f}")
```

## 6. Попробуйте поменять длину шага.

Будет ли сходиться алгоритм, если делать более длинные шаги? Как меняется число итераций при уменьшении длины шага?

```python
for k in [0.001, 0.01, 0.1, 0.5, 1.0]:
    w1_k, w2_k = gradient_descent(X, y, C=0, k=k)
    auc_k = roc_auc_score(y, sigmoid(w1_k * x1 + w2_k * x2))
    print(f"k={k:.3f}: AUC={auc_k:.3f}, w1={w1_k:.4f}, w2={w2_k:.4f}")
```

## 7. Попробуйте менять начальное приближение. Влияет ли оно на что-нибудь?

```python
def gradient_descent_with_init(X, y, w1_init=0.0, w2_init=0.0, C=0, k=0.1, max_iter=10000):
    w1, w2 = w1_init, w2_init
    x1 = X[1].values
    x2 = X[2].values
    for _ in range(max_iter):
        pred = y * (w1 * x1 + w2 * x2)
        error = 1 - sigmoid(pred)
        w1_new = w1 + k * np.mean(y * x1 * error) - k * C * w1
        w2_new = w2 + k * np.mean(y * x2 * error) - k * C * w2
        if np.sqrt((w1_new - w1) ** 2 + (w2_new - w2) ** 2) < 1e-5:
            break
        w1, w2 = w1_new, w2_new
    return w1, w2

for w1_init, w2_init in [(0, 0), (1, 1), (-1, -1), (0.5, -0.5)]:
    w1_i, w2_i = gradient_descent_with_init(X, y, w1_init=w1_init, w2_init=w2_init)
    auc_i = roc_auc_score(y, sigmoid(w1_i * x1 + w2_i * x2))
    print(f"init=({w1_init}, {w2_init}): AUC={auc_i:.3f}, w=({w1_i:.4f}, {w2_i:.4f})")
```
