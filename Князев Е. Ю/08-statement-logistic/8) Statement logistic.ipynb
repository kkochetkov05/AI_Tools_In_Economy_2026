# 1. Реализуйте градиентный спуск для логистической регрессии

Реализуйте обучение модели для двух случаев: без регуляризации и с L2-регуляризацией.


```python
import numpy as np
import pandas as pd
from sklearn.metrics import roc_auc_score

data = pd.read_csv('data-logistic.csv', header=None)
X=data[[1,2]]
y=data[0]

def sigmoid(z):
    return 1/(1+np.exp(-z))
def gradient_descent(X, y, C=0, k=0.1, max_iter=10000):
    w1, w2 = 0.0, 0.0
    x1 = X[1].values
    x2 = X[2].values
    for _ in range(max_iter):
        pred = y * (w1*x1 + w2*x2)
        error = 1-sigmoid(pred)
        w1_new = w1 + k*np.mean(y * x1 * error) - k*C*w1
        w2_new = w2 + k*np.mean(y * x2 * error) - k*C*w2
        if np.sqrt((w1_new - w1)**2 + (w2_new - w2)**2)  <1e-5:
            break

        w1, w2 = w1_new, w2_new
    return w1, w2

w1, w2 = gradient_descent(X, y, C=0)
w1_reg, w2_reg = gradient_descent(X, y, C=10)
x1 = X[1].values
x2 = X[2].values
```

# 2. Оцените качество полученных моделей

Посчитайте ROC-AUC для модели без регуляризации и для модели с регуляризацией.


```python
auc = roc_auc_score(y,sigmoid(w1*x1 + w2*x2))
auc_reg=roc_auc_score(y,sigmoid(w1_reg*x1 + w2_reg*x2))

print(f"{auc:.3f} {auc_reg:.3f}")
```
