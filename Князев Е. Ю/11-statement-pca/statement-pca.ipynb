## 1. Загрузите данные `close_prices.csv`.

В этом файле приведены цены акций 30 компаний на закрытии торгов за каждый день периода.

```python
import numpy as np
import pandas as pd
from sklearn.decomposition import PCA

prices_data = pd.read_csv('close_prices.csv')
X = prices_data.drop(columns=['date'])
```

## 2. На загруженных данных обучите преобразование PCA с числом компоненты равным 10. Скольких компонент хватит, чтобы объяснить 90% дисперсии?

```python
pca = PCA(n_components=10)
pca.fit(X)

explained_variance = pca.explained_variance_ratio_
accumulated_variance = 0
needed_components = 0

for i, variance in enumerate(explained_variance):
    accumulated_variance += variance
    if accumulated_variance >= 0.90:
        needed_components = i + 1
        break

print(f"Количество компонент, объясняющих 90% дисперсии: {needed_components}")
```

## 3. Примените построенное преобразование к исходным данным и возьмите значения первой компоненты.

```python
X_transformed = pca.transform(X)
first_pc = X_transformed[:, 0]
```

## 4. Загрузите информацию об индексе Доу-Джонса из файла `djia_prices.csv`. Чему равна корреляция Пирсона между первой компонентой и индексом Доу-Джонса?

```python
djia_data = pd.read_csv('djia_index.csv')
djia_values = djia_data['^DJI'].values

correlation_matrix = np.corrcoef(first_pc, djia_values)
pearson_corr = correlation_matrix[0, 1]

print(f"Корреляция Пирсона между первой компонентой и индексом: {pearson_corr:.2f}")
```

## 5. Какая компания имеет наибольший вес в первой компоненте?

```python
first_component_weights = pca.components_[0]
max_weight_index = np.argmax(first_component_weights)

company_names = X.columns
best_company = company_names[max_weight_index]

print(f"Компания с наибольшим весом в первой компоненте: {best_company}")
```
