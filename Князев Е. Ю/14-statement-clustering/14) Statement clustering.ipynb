# 1. Подготовьте пиксели изображения

Представьте изображение как набор пикселей и подготовьте его для кластеризации с помощью `KMeans`.


```python
import numpy as np
import pandas as pd
from skimage.io import imread
from skimage import img_as_float
from sklearn.cluster import KMeans

image = imread('parrots.jpg')
image = img_as_float(image)
pixels = image.reshape((-1, 3))

def psnr(img1, img2):
    mse = ((img1 - img2) ** 2).mean()
    return 100 if mse == 0 else 10 * np.log10(1.0 / mse)
```

# 2. Найдите минимальное число кластеров

Для каждого `k` постройте две версии восстановления: по среднему и по медиане цвета внутри кластера. Выберите минимальное `k`, при котором `PSNR > 20`.


```python
for k in range(1, 21):
    kmeans = KMeans(n_clusters=k, random_state=241, init='k-means++')
    labels = kmeans.fit_predict(pixels)
    data=pd.DataFrame(pixels)
    data['label'] = labels

    means = data.groupby('label').mean().values
    medians = data.groupby('label').median().values

    image_mean = means[labels].reshape(image.shape)
    image_median = medians[labels].reshape(image.shape)

    psnr_mean = psnr(image, image_mean)
    psnr_median = psnr(image, image_median)

    if max(psnr_mean, psnr_median) > 20:
        print(k)
        break
```
