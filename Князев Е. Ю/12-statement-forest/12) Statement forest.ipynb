# 1. Подготовьте данные

Преобразуйте столбец `Sex` в числовой признак и разделите данные на признаки и целевую переменную.


```python
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score
from sklearn.model_selection import KFold, cross_val_score

data =pd.read_csv('abalone.csv')
data['Sex'] = data['Sex'].map(lambda x:1 if x=='M' else (-1 if x=='F' else 0))
X=data.drop(columns=['Rings'])
Y=data['Rings']


kf = KFold(n_splits=5, shuffle=True, random_state=1)
results = []
```

# 2. Подберите размер случайного леса

Переберите количество деревьев от 1 до 50 и найдите минимальное значение, при котором среднее `R2` превышает 0.52.


```python
for i in range(1,51):
    model = RandomForestRegressor(n_estimators=i, random_state=1)
    scores = cross_val_score(model, X, Y, cv=kf, scoring='r2')
    mean_score = scores.mean()
    results.append(mean_score)
    print(f'n_estimators={i}, R2={mean_score:.4f}')
for n, score in enumerate(results, start=1):
    if score > 0.52:
        print(f'\nОтвет: минимальное количество деревьев = {n}')
        break
```
