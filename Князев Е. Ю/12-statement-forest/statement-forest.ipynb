## 1. Загрузите данные из файла `abalone.csv`.

Это датасет, в котором требуется предсказать возраст ракушки (число колец) по физическим измерениям.

```python
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score
from sklearn.model_selection import KFold, cross_val_score

data = pd.read_csv('abalone.csv')
```

## 2. Преобразуйте признак `Sex` в числовой: значение F должно перейти в -1, I — в 0, M — в 1.

Если вы используете Pandas, то подойдет следующий код: `data['Sex'] = data['Sex'].map(lambda x: 1 if x == 'M' else (-1 if x == 'F' else 0))`

```python
data['Sex'] = data['Sex'].map(lambda x: 1 if x == 'M' else (-1 if x == 'F' else 0))
```

## 3. Разделите содержимое файлов на признаки и целевую переменную.

В последнем столбце записана целевая переменная, в остальных — признаки.

```python
X = data.drop(columns=['Rings'])
Y = data['Rings']
```

## 4. Обучите случайный лес (`sklearn.ensemble.RandomForestRegressor`) с различным числом деревьев: от 1 до 50 (`random_state=1`).

Для каждого из вариантов оцените качество работы полученного леса на кросс-валидации по 5 блокам. Используйте параметры `random_state=1` и `shuffle=True` при создании генератора кросс-валидации. В качестве меры качества воспользуйтесь коэффициентом детерминации (`sklearn.metrics.r2_score`).

```python
kf = KFold(n_splits=5, shuffle=True, random_state=1)
results = []

for i in range(1, 51):
    model = RandomForestRegressor(n_estimators=i, random_state=1)
    scores = cross_val_score(model, X, Y, cv=kf, scoring='r2')
    mean_score = scores.mean()
    results.append(mean_score)
    print(f'n_estimators={i}, R2={mean_score:.4f}')
```

## 5. Определите, при каком минимальном количестве деревьев случайный лес показывает качество на кросс-валидации выше 0.52. Это количество и будет ответом на задание.

```python
for n, score in enumerate(results, start=1):
    if score > 0.52:
        print(f'Ответ: минимальное количество деревьев = {n}')
        break
```
