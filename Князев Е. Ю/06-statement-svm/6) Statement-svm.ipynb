# 1. Обучите SVM с линейным ядром

Используйте `SVC(C=100000, kernel="linear", random_state=241)` и найдите индексы опорных объектов.


```python
import pandas as pd
from sklearn.svm import SVC

data = pd.read_csv('svm-data.csv', header=None)
X=data[[1,2]]
y=data[0]

clf = SVC(C=100000, kernel='linear', random_state=241)
clf.fit(X, y)

support_indices = sorted(clf.support_ + 1)
print(','.join(map(str, support_indices)))
```
