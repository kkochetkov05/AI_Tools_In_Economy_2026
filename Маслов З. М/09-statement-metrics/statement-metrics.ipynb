1. Загрузите файл classification.csv. В нем записаны истинные классы
объектов выборки (колонка true) и ответы некоторого классификатора (колонка predicted).


```python
import pandas as pd
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, precision_recall_curve

classification = pd.read_csv('./classification.csv')
scores = pd.read_csv('./scores.csv')
```

2. Заполните таблицу ошибок классификации:


```python
TP = len(classification[(classification['true'] == classification['pred']) & (classification['true'] == 1)])
FP = len(classification[(classification['true'] != classification['pred']) & (classification['true'] == 0)])
TN = len(classification[(classification['true'] == classification['pred']) & (classification['true'] == 0)])
FN = len(classification[(classification['true'] != classification['pred']) & (classification['true'] == 1)])

error_table = pd.DataFrame(data=[[TP, FP], [TN, FN]], columns=['Actual Positive', 'Actual Negative'])
print(error_table)
```

3. Посчитайте основные метрики качества классификатора:
• Accuracy (доля верно угаданных) — sklearn.metrics.accuracy
• Precision (точность) — sklearn.metrics.accuracy.precision_score
• Recall (полнота) — sklearn.metrics.recall_score
• F-мера — sklearn.metrics.f1_score

```python
print(accuracy_score(classification['true'], classification['pred']))
print(precision_score(classification['true'], classification['pred']))
print(recall_score(classification['true'], classification['pred']))
print(f1_score(classification['true'], classification['pred']))
```

5. Посчитайте площадь под ROC-кривой для каждого классификатора. Какой классификатор имеет наибольшее значение метрики
AUC-ROC (укажите название столбца с ответами этого классификатора)? 

```python
auc_roc_scores = pd.DataFrame(data = None, columns=scores.keys()[1:])
for i, key in enumerate(auc_roc_scores.columns.tolist()):
    auc_roc_scores[key] = pd.Series(roc_auc_score(scores['true'], scores.iloc[:, i+1]))
    print(roc_auc_score(scores['true'], scores.iloc[:, i+1]))
print(auc_roc_scores.iloc[0].idxmax())
```

6. Какой классификатор достигает наибольшей точности (Precision)
при полноте (Recall) не менее 70% (укажите название столбца с ответами этого классификатора)?

```python
pr_scores = {}
for i in scores[:, 1:]:
    curve = precision_recall_curve(scores['true'], scores[i])
    df = pd.DataFrame({'precision': curve[0], 'recall': curve[1]})
    pr_scores[i] = df[df['recall'] >= 0.7]['precision'].max()
print(pd.Series(pr_scores).idxmax(), pd.Series(pr_scores).max())
```
