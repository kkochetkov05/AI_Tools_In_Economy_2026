1. Загрузите объекты из новостного датасета 20 newsgroups, относящиеся к категориям "космос"и "атеизм"(инструкция приведена выше).


```python
from sklearn import datasets
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from  sklearn.model_selection import GridSearchCV, KFold, cross_val_score
import numpy as np

texts = datasets.fetch_20newsgroups(
    subset='all',
    categories=['alt.atheism', 'sci.space']
)
```

2. Вычислите TF-IDF-признаки для всех текстов.

```python
transform_X = TfidfVectorizer()
y = texts.target
X = transform_X.fit_transform(texts.data, y)
```

3. Подберите минимальный лучший параметр C из множества [10−5
, 10−4
, ...104
, 105
]
для SVM с линейным ядром (kernel=’linear’) при помощи кроссвалидации по 5 блокам.

```python
grid = {'C': [np.power(10.0, i) for i in np.arange(-5, 7)]}
clf = SVC(kernel='linear', random_state=241)
cv = KFold(n_splits=5, shuffle=True, random_state=241)
grid = GridSearchCV(estimator=clf, cv=cv, param_grid=grid, scoring='accuracy')
grid.fit(X, y)
print(grid.best_estimator_)
```

4. Обучите SVM по всей выборке с лучшим параметром C, найденным
на предыдущем шаге.

```python
best_model = grid.best_estimator_
```

5. Найдите 10 слов с наибольшим по модулю весом. Они являются
ответом на это задание. 

```python
coefs = np.abs(best_model.coef_.toarray().flatten())
idx_top = np.argsort(coefs)[-10:]
features = transform_X.get_feature_names_out()
print(",".join(sorted([features[i] for i in idx_top])))
```
