## 3. Реализуйте градиентный спуск для обычной и L2-регуляризованной (с коэффициентом регуляции 10) логистической регрессии. Используйте длину шага k=0.1. В качестве начального приближения используйте вектор (0, 0).

## 4. Запустите градиентный спуск и доведите до сходимости (евклидово расстояние между векторами весов на соседних итерациях должно быть не больше 1e-5). Рекомендуется ограничить сверху число итераций десятью тысячами.

```python
def GD_iter(W, l, X, y, k, C):
    return W + k/l * sum([y[i]*X[i] * (1 - 1/(1 + np.exp(-y[i]* W @ X[i]))) for i in range(l)]) - k*C*W

def GD(W, k, eps = 10**(-5), C=0, max_iter = 10**4, X = data_np[:, 1:], y = data_np[:, 0]):
    l = len(y)
    for _ in range(max_iter):
        W_new = GD_iter(W, l, X, y, k, C)
        if np.linalg.norm(W-W_new) < eps:
            return W_new
        W = W_new
    return W_new

def sigm(x, W):
    return 1/(1+ np.exp(-W@x))
def predict(X, W):
    return np.array([sigm(X[i, :], W) for i in range(len(X))])
```

```python
def evaluate(w1, w2, k, C):
    W = np.array([w1,w2])
    W_best = GD(W, 0.1)
    W_best_reg = GD(W, 0.1, C=C)
    return float(roc_auc_score(y, predict(X, W_best))), float(roc_auc_score(y, predict(X, W_best_reg)))
```
