# Лабораторная работа 1: анализ данных Titanic

```python
import numpy as np
import pandas as pd
```

```python
data = pd.read_csv('titanic.csv')
data.head()
```

# 1. Какое количество мужчин и женщин ехало на корабле?

В качестве ответа приведите два числа через пробел.

```python
ans_1 = (data.loc[data.Sex == 'male', 'Sex'].count(), data.loc[data.Sex == 'female', 'Sex'].count())

with open('1.txt', 'w') as file:
    file.write(f'{ans_1[0]} {ans_1[1]}')
```

# 2. Какой части пассажиров удалось выжить?

Посчитайте долю выживших пассажиров. Ответ приведите в процентах (число в интервале от 0 до 100, знак процента не нужен), округлив до двух знаков.

```python
survived_count = len(data.loc[data.Survived == 1])
ans_2 = (survived_count / len(data)) * 100

with open('2.txt', 'w') as file:
    file.write(f'{ans_2:.2f}')
```

# 3. Какую долю пассажиры первого класса составляли среди всех пассажиров?

Ответ приведите в процентах (число в интервале от 0 до 100, знак процента не нужен), округлив до двух знаков.

```python
ans_3 = data.loc[data.Pclass == 1, 'Pclass'].count()/len(data) * 100

with open('3.txt', 'w') as file:
    file.write(f'{ans_3:.2f}')
```

# 4. Какого возраста были пассажиры?

Посчитайте среднее и медиану возраста пассажиров. В качестве ответа приведите два числа через пробел (среднее и медиану), округлив каждое до двух знаков.

```python
ans_4 = data.Age.mean(), data.Age.median()

with open('4.txt', 'w') as file:
    file.write(f'{ans_4[0]:.2f} {ans_4[1]:.2f}')
```

# 5. Коррелируют ли число братьев/сестер с числом родителей/детей?

Посчитайте корреляцию Пирсона между признаками SibSp и Parch. Ответ округлите до двух знаков.

```python
ans_5 = data['SibSp'].corr(data['Parch'])

with open('5.txt', 'w') as file:
    file.write(f'{ans_5:.2f}')
```

# 6. Какое самое популярное женское имя на корабле?

Извлеките из полного имени пассажира (колонка Name) его личное имя (First Name). Это задание — типичный пример того, с чем сталкивается специалист по анализу данных. Данные очень разнородные и шумные, но из них требуется извлечь необходимую информацию. Попробуйте вручную разобрать несколько значений столбца Name и выработать правило для извлечения имен, а также разделения их на женские и мужские.

В качестве ответа укажите одно имя (строка).

```python
def find_name(s):
    words = s.split()
    if words[1] == 'Mrs.':
        for w in words:
            if w[0] == '(':
                return w[1:-1] if w[-1] == ')' else w[1:]
    elif words[1] == 'Miss.':
        return words[2]
    return None

def popular_name_woman(df):
    name_series = df.Name.apply(find_name).dropna()
    pop_name = name_series.value_counts().idxmax()
    return pop_name
    
ans_6 = popular_name_woman(data)

with open('6.txt', 'w') as file:
    file.write(f'{ans_6}')
```
