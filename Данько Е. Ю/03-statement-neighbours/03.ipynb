# Лабораторная работа 3: классификация вина методом k ближайших соседей

## 1. Загрузите выборку Wine по адресу https://archive.ics.uci.edu/ml/machine-learning-databases/wine/wine.data

```python
import pandas as pd
from sklearn.model_selection import KFold
from sklearn.preprocessing import scale
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import cross_val_score

import seaborn as sb
import matplotlib.pyplot as plt
```

```python
col_names = ["Class", "Alcohol", "Malic acid", "Ash", "Alcalinity of ash", "Magnesium", "Total phenols","Flavanoids","Nonflavanoid phenols","Proanthocyanins","Color intensity","Hue","OD280/OD315 of diluted wines", "Proline"]
data = pd.read_csv('wine.data', header=None, names=col_names)
data
```

## 2. Извлеките из данных признаки и классы. Класс записан в первом столбце (три варианта), признаки — в столбцах со второго по последний.

```python
X = data[col_names[1:]]
y = data['Class']
```

## 3. Оценку качества необходимо провести методом кросс-валидации по 5 блокам (5-fold). Создайте генератор разбиений, который перемешивает выборку перед формированием блоков (shuffle=True). Для воспроизводимости результата, создавайте генератор KFold с фиксированным параметром random_state=42. В качестве меры качества используйте долю верных ответов (accuracy).

```python
kf = KFold(shuffle=True, n_splits=5, random_state=42)

def evaluate_KNN(k, X, y):
    knn = KNeighborsClassifier(n_neighbors=k)
    return cross_val_score(knn, X, y, cv = kf, scoring='accuracy').mean()

eval_df = pd.DataFrame(columns=['k', 'eval_mean'])

for k in range(1, 51):
    eval_mean = evaluate_KNN(k, X, y)
    eval_df.loc[len(eval_df)] = [k, eval_mean]
```

## 4. Найдите точность классификации на кросс-валидации для метода k ближайших соседей (sklearn.neighbors.KNeighborsClassifier), при k от 1 до 50. При каком k получилось оптимальное качество? Чему оно равно (число в интервале от 0 до 1)? Данные результаты и будут ответами на вопросы 1 и 2.

```python
ans1 = eval_df.loc[eval_df['eval_mean'].idxmax()]

with open('1.txt', 'w') as file:
    file.write(f'{int(ans1['k'])}')
with open('2.txt', 'w') as file:
    file.write(f'{ans1['eval_mean']:.2f}')
```

## 5. Произведите масштабирование признаков с помощью функции sklearn.preprocessing.scale. Снова найдите оптимальное k на кросс-валидации.

```python
X_scl = scale(X)


eval_df_scl = pd.DataFrame(columns=['k', 'eval_mean'])

for k in range(1, 51):
    eval_mean = evaluate_KNN(k, X_scl, y)
    eval_df_scl.loc[len(eval_df_scl)] = [k, eval_mean]

ans2 = eval_df_scl.loc[eval_df_scl['eval_mean'].idxmax()]

with open('3.txt', 'w') as file:
    file.write(f'{int(ans2['k'])}')
with open('4.txt', 'w') as file:
    file.write(f'{ans2['eval_mean']:.2f}')
```
