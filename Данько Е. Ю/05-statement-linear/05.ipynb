# Лабораторная работа 5: персептрон и нормализация данных

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import Perceptron
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score
```

## 1. Загрузите обучающую и тестовую выборки из файлов perceptron-train.csv и perceptron-test.csv. Целевая переменная записана в первом столбце, признаки — во втором и третьем.

```python
col_names = ['y', 'f1', 'f2']
data_train = pd.read_csv('perceptron-train.csv', header=None, names=col_names)
data_test = pd.read_csv('perceptron-test.csv', header=None, names=col_names)
```

```python
X_train, y_traint = data_train[col_names[1:]], data_train['y']
X_test, y_test = data_test[col_names[1:]], data_test['y']
```

Вспомогательная функция для обучения и оценки точности.

```python
def eval_perceptron(mdl, X_t, y_t, X, y):
    mdl_fit = mdl.fit(X_t, y_t)
    y_p = mdl.predict(X)
    return accuracy_score(y, y_p)
```

## 2. Обучите персептрон со стандартными параметрами и random_state=241.

```python
percept_model = Perceptron(random_state=241)
```

## 3. Подсчитайте качество (долю правильно классифицированных объектов, accuracy) полученного классификатора на тестовой выборке. (Сохраняется в переменной acc1)

```python
acc1 = eval_perceptron(percept_model, X_train, y_traint, X_test, y_test)
```

## 4. Нормализуйте обучающую и тестовую выборку с помощью класса StandardScaler.

```python
std_scale = StandardScaler().fit(X_train)
X_train_std = std_scale.transform(X_train)
X_test_std = std_scale.transform(X_test)
```

## 5. Обучите персептрон на новых выборках. Найдите долю правильных ответов на тестовой выборке. (Результат — acc2)

```python
acc2 = eval_perceptron(percept_model, X_train_std, y_traint, X_test_std, y_test)
```

## 6. Найдите разность между качеством на тестовой выборке после нормализации и качеством до нее. Это число и будет ответом на задание.

```python
with open('1.txt', 'w') as file:
    file.write(f'{abs(acc2-acc1):.3f}')
```
