# Лабораторная работа 9: оценка качества классификации

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb
from  sklearn.metrics import accuracy_score, recall_score, precision_score, f1_score, roc_auc_score, precision_recall_curve
```

## 1. Загрузите файл classification.csv. В нем записаны истинные классы объектов выборки (колонка true) и ответы некоторого классификатора (колонка predicted).

```python
data_clf = pd.read_csv('classification.csv')
data_clf
```

## 2. Заполните таблицу ошибок классификации: подсчитайте величины TP, FP, FN и TN. Ответ — четыре числа через пробел.

```python
TP = len(data_clf[(data_clf['true'] == 1) & (data_clf['pred'] == 1)])
TN = len(data_clf[(data_clf['true'] == 0) & (data_clf['pred'] == 0)])
FP = len(data_clf[(data_clf['true'] == 0) & (data_clf['pred'] == 1)])
FN = len(data_clf[(data_clf['true'] == 1) & (data_clf['pred'] == 0)])

ans_1 = TP, FP, FN, TN

with open('1.txt', 'w') as file:
    file.write(f'{TP} {FP} {FN} {TN}')
```

## 3. Посчитайте основные метрики качества классификатора: Accuracy, Precision, Recall, F-measure. Ответ — четыре числа через пробел (округлённые до двух знаков).

```python
accuracy = accuracy_score(data_clf['true'], data_clf['pred'])
precision = precision_score(data_clf['true'], data_clf['pred'])
recall = recall_score(data_clf.true, data_clf.pred)
f1 = f1_score(data_clf.true, data_clf.pred)

ans_2 = accuracy, precision, recall, f1
with open('2.txt', 'w') as file:
    file.write(f'{accuracy:.2f} {precision:.2f} {recall:.2f} {f1:.2f}')
```

## 4. Загрузите файл scores.csv. В нем записаны истинные классы и значения степени принадлежности положительному классу для четырёх классификаторов: score_logreg, score_svm, score_knn, score_tree.

```python
data_score = pd.read_csv('scores.csv')
data_score
```

## 5. Посчитайте площадь под ROC-кривой для каждого классификатора. Какой классификатор имеет наибольшее значение метрики AUC-ROC? Укажите название столбца с ответами этого классификатора.

```python
clm_name = list(data_score.columns)[1:]

dict_roc_auc = {name: roc_auc_score(data_score.true, data_score.loc[:, name]) for name in clm_name}
ans_3 = max(dict_roc_auc, key=dict_roc_auc.get)

with open('3.txt', 'w') as file:
    file.write(f'{ans_3}')
```
