### 1. Загрузите данные close_prices.csv. 
В этом файле приведены цены
акций 30 компаний на закрытии торгов за каждый день периода.

```python
import pandas as pd

df = pd.read_csv('close_prices.csv')

X = df.drop(columns = ['date'])
```

### 2. На загруженных данных обучите преобразование PCA с числом компоненты равным 10. 
Скольких компонент хватит, чтобы объяснить 90% дисперсии?

```python
import numpy as np
from sklearn.decomposition import PCA

pca = PCA(n_components = 10)

pca.fit(X)

variance_cumsum = np.cumsum(pca.explained_variance_ratio_)

print(np.where(variance_cumsum >= 0.9)[0][0]+1)
```

### 3. Примените построенное преобразование к исходным данным и возьмите значения первой компоненты.

```python
X_pca = pca.transform(X)
first_component = X_pca[:,0]
```

### 4. Загрузите информацию об индексе Доу-Джонса из файла djia_prices.csv.
Чему равна корреляция Пирсона между первой компонентой и индексом Доу-Джонса?

```python
djia = pd.read_csv('djia_index.csv')
djia_vals = djia['^DJI'].values

print(np.corrcoef(first_component, djia_vals)[0][1].round(2))
```

### 5. Какая компания имеет наибольший вес в первой компоненте?

```python
weights = pd.Series(pca.components_[0], index = X.columns)
print(weights.abs().idxmax())
```
