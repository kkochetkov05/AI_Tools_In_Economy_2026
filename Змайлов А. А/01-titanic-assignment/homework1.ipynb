### 6. Какое самое популярное женское имя на корабле? 
Извлеките из полного имени пассажира (колонка Name) его личное имя (First
Name).

```python
import re

def get_first_name(full_name):
    
    s = str(full_name)

    m_title = re.search(r',\s*([^.]+)\.', s)
    title = m_title.group(1).strip() if m_title else None

    if title == 'Mrs':
        m_paren = re.search(r'\(([^)]+)\)', s)
        if m_paren:
            s = m_paren.group(1)
        else:
            s = re.sub(r'^.*?\.\s*', '', s)  # после титула
    else:
        s = re.sub(r'^.*?\.\s*', '', s)      # после титула

    m_name = re.search(r"[A-Za-z]+(?:-[A-Za-z]+)?", s)
    return m_name.group(0) if m_name else None

titanic['FirstName'] = titanic['Name'].apply(get_first_name)

top_female_name = titanic.loc[titanic['Sex'] == 'female', 'FirstName'].value_counts().idxmax()
print(top_female_name)
```
